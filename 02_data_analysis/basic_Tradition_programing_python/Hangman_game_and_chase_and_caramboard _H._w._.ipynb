Hangman game


```python

import turtle
import random

# Random words ki list
words = ["python", "turtle", "computer", "coding"]

# Ek random word choose karna
word = random.choice(words)

# Guessed letters store karna
letters = []

# Wrong guesses count karna
wrong = 0

# Turtle setup
t = turtle.Turtle()
t.speed(10)
t.pensize(3)

# Hangman stand banana
t.penup()
t.goto(-200, -200)
t.pendown()
t.forward(400)

t.left(90)
t.forward(300)
t.right(90)
t.forward(120)
t.right(90)
t.forward(40)

# Game start
while wrong < 6:

    # Word ke guessed letters dikhana
    show = ""

    for x in word:
        if x in letters:
            show += x + " "
        else:
            show += "_ "

    print("\nWord:", show)
    print("Wrong guesses:", wrong)

    # Word guess ho gaya toh game jeet gaye
    if "_" not in show:
        print("You Won!")
        break

    # User se ek letter lena
    guess = input("Guess a letter: ").lower()

    # Sirf ek letter allow karna
    if len(guess) != 1 or not guess.isalpha():
        print("Please enter one letter!")
        continue

    # Repeat guess check karna
    if guess in letters:
        print("Already guessed!")
        continue

    # Letter ko list mein add karna
    letters.append(guess)

    # Sahi ya galat guess check karna
    if guess in word:
        print("Correct!")
        continue

    print("Wrong!")
    wrong += 1

    # Head banana
    if wrong == 1:
        t.penup()
        t.goto(20, 150)
        t.pendown()
        t.circle(30)

    # Body banana
    elif wrong == 2:
        t.penup()
        t.goto(20, 150)
        t.setheading(270)
        t.pendown()
        t.forward(100)

    # Left hand banana
    elif wrong == 3:
        t.penup()
        t.goto(20, 120)
        t.setheading(210)
        t.pendown()
        t.forward(50)

    # Right hand banana
    elif wrong == 4:
        t.penup()
        t.goto(20, 120)
        t.setheading(330)
        t.pendown()
        t.forward(50)

    # Left leg banana
    elif wrong == 5:
        t.penup()
        t.goto(20, 50)
        t.setheading(240)
        t.pendown()
        t.forward(60)

    # Right leg banana
    elif wrong == 6:
        t.penup()
        t.goto(20, 50)
        t.setheading(300)
        t.pendown()
        t.forward(60)

# Game haarne par
if wrong == 6:
    print("\nGame Over!")
    print("Correct word was:", word)

# Turtle window open rakhna
turtle.done()
```

Carrom Board Game 

```python

import random

# Game ka title print kar rahe hain
print("===== CARROM BOARD GAME =====")

# Player ka naam input le rahe hain
name = input("Apna naam likho: ")

# Starting score zero rakha hai
score = 0

# Board par total 10 coins hain
coins = 10

# Jab tak coins khatam nahi hote, game chalega
while coins > 0:

    # Player ko remaining coins dikha rahe hain
    print("\nCoins remaining:", coins)

    # Player se shot lene ke liye bol rahe hain
    input("Shot marne ke liye Enter dabao...")

    # Random number se shot ka result decide hoga
    shot = random.randint(1, 3)

    # Agar result 1 ya 2 hai, coin pocket hoga
    if shot != 3:
        print("Great shot! Coin pocket ho gaya!")

        # Score mein 10 points add honge
        score = score + 10

        # Board se ek coin kam hoga
        coins = coins - 1

    # Agar result 3 hai, shot miss hoga
    else:
        print("Shot miss ho gaya! Dobara try karo.")

# Game khatam hone par final score dikhayenge
print("\n===== GAME OVER =====")

# Player ka naam aur final score print karenge
print("Player:", name)
print("Final Score:", score)
print("Pocketed Coins:", 10 - coins)
```
