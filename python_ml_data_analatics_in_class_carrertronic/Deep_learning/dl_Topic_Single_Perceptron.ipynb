GRADIENT DESCENT — MY UNDERSTANDING

GOAL:
Cost/Loss ko minimum karna hai.
Minimum point ke paas gradient approximately 0 hota hai.

MODEL:
Prediction = w × x + b

w = weight / slope
b = bias / intercept

VARIABLES:
w_old  → current weight
b_old  → current bias
h      → very small change, gradient estimate karne ke liye
alpha  → learning rate / step size
i      → iteration count


MAIN PROCESS:

1. Current w aur b se prediction karo.

2. Current cost calculate karo.

3. w ko thoda change karo:
   w + h

4. New cost calculate karo.

5. Gradient w ke respect mein:
   df_dw = (new cost - old cost) / h

6. w update:
   w_new = w_old - alpha × df_dw


7. Ab b ke saath same process:

   b + h

8. New cost calculate karo.

9. Gradient b ke respect mein:
   df_db = (new cost - old cost) / h

10. b update:
    b_new = b_old - alpha × df_db


11. Check:

    df_dw ≈ 0
    AND
    df_db ≈ 0

    → STOP


12. Agar zero nahi hai:

    w_old = w_new
    b_old = b_new

    → next iteration


IMPORTANT IDEA:

Gradient batata hai cost kis direction mein increase ho rahi hai.

Gradient Descent uske opposite direction mein move karta hai.

Isliye:

w_new = w_old - alpha × df_dw
b_new = b_old - alpha × df_db


ONE-LINE MEMORY:

Cost ↓
→ Gradient find
→ Opposite direction mein w,b update
→ Repeat
→ Gradient ≈ 0
→ Minimum ke paas


FINITE DIFFERENCE IDEA:

Parameter ko bahut chhota change karo
→ cost ka difference dekho
→ difference / h
→ approximate gradient


PYTORCH CONNECTION:

Manual:
gradient calculate → parameter update

PyTorch:
loss.backward() → optimizer.step()

CONCEPT SAME HAI.
Framework calculation ko automate karta hai.

```python
# Starting mein weight ki value 100 rakhi
w_old = 100

# Starting mein bias ki value 11 rakhi
b_old = 11

# Gradient calculate karne ke liye bahut chhota change
h = 0.00001

# Learning rate: weight aur bias ko kitna update karna hai
alpha = 0.01


# Iteration count 0 se start kiya
i = 0


# Ye loop baar-baar chalega jab tak gradient approximately 0 nahi ho jata
while True:

    # w ko thoda sa (+h) change karke nayi cost calculate kar rahe hain
    f_wb_h = sum(
        (data['Salary'] -
         ((w_old + h) * data['YearsExperience'] + b_old)) ** 2
    ) / len(data)

    # Current/original w aur b ke saath current cost calculate kar rahe hain
    f_wb = sum(
        (data['Salary'] -
         ((w_old) * data['YearsExperience'] + b_old)) ** 2
    ) / len(data)


    # w ko thoda change karne par cost kitni change hui,
    # usse w ka gradient calculate kar rahe hain
    df_dw = (f_wb_h - f_wb) / h


    # Gradient ke according w ko update kar rahe hain
    w_new = w_old - alpha * df_dw


    # Ab b ko thoda sa (+h) change karke nayi cost calculate kar rahe hain
    f_wb_bh = sum(
        (data['Salary'] -
         ((w_old) * data['YearsExperience'] + (b_old + h))) ** 2
    ) / len(data)

    # Current/original w aur b ke saath current cost dobara calculate kar rahe hain
    f_wb = sum(
        (data['Salary'] -
         ((w_old) * data['YearsExperience'] + b_old)) ** 2
    ) / len(data)


    # b ko thoda change karne par cost kitni change hui,
    # usse b ka gradient calculate kar rahe hain
    df_db = (f_wb_bh - f_wb) / h


    # Gradient ke according b ko update kar rahe hain
    b_new = b_old - alpha * df_db


    # Check kar rahe hain ki w aur b dono ke gradients approximately 0 hain ya nahi
    if round(df_dw, 6) == 0 and round(df_db, 6) == 0:

        # Agar dono gradient approximately 0 hain,
        # to learning/loop ko stop kar do
        break

    else:

        # Gradient 0 nahi hai, isliye ek aur iteration count karo
        i = i + 1

        # Naya w ab next iteration ka current/old w ban gaya
        w_old = w_new

        # Naya b ab next iteration ka current/old b ban gaya
        b_old = b_new


# Final w ka gradient aur b ka gradient print kar rahe hain
print(round(df_dw, 6), round(df_db, 6))

# Total iterations aur final w,b ki values print kar rahe hain
print(i, w_new, b_new)
```
