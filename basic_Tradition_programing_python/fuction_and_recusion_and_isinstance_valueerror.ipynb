```python
# <!-- Practice Question 1 — Student Marks Validator
# Ek function check_marks(name, marks) banao jo:
# 1. Check kare ki name string hai.
# 2. Check kare ki marks ek list hai.
# 3. List ke har element ka type int ya float hai. True aur False allowed nahi hain.
# 4. Agar koi data type galat ho, toh raise TypeError use kare.
# 5. Agar saara data sahi ho, toh marks ka total return kare. -->


def ceck_marks(name, marks):  # Function bana rahe hain, jo name aur marks lega

    if not isinstance(name, str):  # Check kar rahe hain ki name string hai ya nahi
        raise TypeError("Name string hona chahiye")  # Galat type hua toh error aayega

    if not isinstance(marks, list):  # Check kar rahe hain ki marks ek list hai ya nahi
        raise TypeError("Marks list honi chahiye")  # List nahi hui toh error aayega

    total = 0  # Total ko 0 se start kar rahe hain

    for i in marks:  # List ke har mark ko ek-ek karke i mein lenge

        if isinstance(i, bool):  # Check kar rahe hain ki value True ya False toh nahi
            raise TypeError("True aur False allowed nahi hain")  # Boolean hua toh error aayega

        if not isinstance(i, (int, float)):  # Check kar rahe hain ki mark int ya float hai
            raise TypeError("Marks sirf int ya float hone chahiye")  # Galat type hua toh error

        total = total + i  # Har valid mark ko total mein add kar rahe hain

    return total  #Saare marks add hone ke baad total return kar rahe hain


print(ceck_marks("Param", [80, 75, 90]))  #Function call karke result print kar rahe hain
```

```python
# Practice Question 1 — Student Age Validator
# Ek function check_age(name, age) banao.
# 1. name ka type str check karna hai.
# 2. age ka type int check karna hai.
# 3. Agar type galat ho, toh raise TypeError use karna hai.
# 4. True aur False ko age ke liye allow nahi karna.
# 5. Agar sab sahi hai, toh return se message return karna hai: "Rahul is 20 years old"

def check_age(name,age):
    if not isinstance(name,str):
        raise TypeError('name sirf str hi ho sakta hai')

    if not isinstance(age,int):
        raise TypeError('ye sirf int type ka ho sakta hai')
    
    if  isinstance(age,bool):
        raise TypeError('bhool are not allowed ya bolean are not allowed')

    return name,'is',age,'years old'

print(check_age('Param_cg_wala',21))
```

```python

# Practice Question 2 - Number Checker

# Ek function check_number(num) banao.

# 1. Check karo ki num ka type int hai.
# 2. Agar type galat ho, toh raise TypeError use karo.
# 3. True aur False ko allow nahi karna.
# 4. Agar sab sahi hai, toh return num * 2 karo.

# Example:
# print(check_number(5))

# Expected Output:
# 10

def check_number(num):  # Function bana rahe hain jo num naam ka input lega

    # Pehle check kar rahe hain ki num True ya False toh nahi hai
    if isinstance(num, bool):
        raise TypeError("Boolean values allowed nahi hain")  # Bool hua toh error dekar rok do

    # Check kar rahe hain ki num integer type ka hai ya nahi
    if not isinstance(num, int):
        raise TypeError("num sirf int hona chahiye")  # Int nahi hua toh error de do

    # Agar dono checks pass ho gaye, toh number ko 2 se multiply karke return karo
    return num * 2


# Function ko 5 dekar call kar rahe hain aur result print kar rahe hain
print(check_number(5))  # Output: 10
```
