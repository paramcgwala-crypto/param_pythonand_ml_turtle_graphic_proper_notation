Level 14: Merging, Joining & Relational-Style Operations

```python
import pandas as pd 
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 261. Create a state-level summary DataFrame containing student count and average Marks.

df_261 = df.groupby('state').agg({'Marks':['count','mean']}).reset_index()
print(df_261)
```

```python
# 262. Create a city-level summary DataFrame containing student count and average Marks.


df_262 = df.groupby('city').agg({'Marks':['count','mean']}).reset_index()
print(df_262)
```

```python
# 263. Merge the state-level summary back into the original DataFrame.

df_263 = df.groupby('state').agg({'Marks':'count'}).reset_index()

df263 = df.merge(df_263,on = 'state',how = 'inner')   # on mai maine bataya hai ki kiske level pay krna hai merge and how mai komn sa join krna hai bataya hai 

print(df263)
```

```python
# 264. Merge the city-level summary back into the original DataFrame.

df_264 = df.groupby('city').agg({'Marks':'count'}).reset_index()
df264 = df.merge(df_264,on= 'city',how = 'inner')
print(df264)
```

```python
# 265. Merge a centre-level summary back into the original DataFrame.

df_265 = df.groupby('center_name').agg({'Marks':'count'}).reset_index()
df265 = df.merge(df_265,on= 'center_name',how = 'inner')
print(df265)
```

```python
# 266. Join state-level statistics to each student record.

df_266 = df.groupby('state').agg({'Marks':'sum'}).reset_index()    # Pehle state-wise statistics (jaise total Marks) nikalo.
df266 = df.merge(df_266,on = 'state' , how = 'left')              # Phir state ke basis par har student ke record ke saath statistics join karo.# Left join se original DataFrame ke saare student records preserve rehte hain.
print(df266)                                                       
```

```python
# 267. Join city-level statistics to each student record.

df_267 = df.groupby('city').agg({'Marks':'sum'}).reset_index()  # Pehle city-wise total Marks nikalo.

df267 = df.merge(df_267, on='city', how='left')  # City ke basis par har student ke record ke saath statistics join karo.
# Left join se original DataFrame ke saare student records preserve rehte hain.

print(df267)
```

```python
# 68. Create separate DataFrames for state totals and state averages, then merge them.

state_total = df.groupby('state').agg({'Marks': 'sum'}).reset_index()  # Har state ke total Marks nikale.

state_average = df.groupby('state').agg({'Marks': 'mean'}).reset_index()  # Har state ke average Marks nikale.

df_268 = state_total.merge(state_average, on='state', how='inner')  # Dono DataFrames ko state ke basis par merge kiya.

print(df_268)  
```

```python
# 269. Create separate DataFrames for city totals and city averages, then merge them.

city_total = df.groupby('city').agg({'Marks': 'sum'})  # Har city ke total Marks nikale.

city_average = df.groupby('city').agg({'Marks': 'mean'})  # Har city ke average Marks nikale.

df_269 = city_total.merge(city_average, on='city', how='inner')  # Total aur average ko city ke basis par merge kiya.

print(df_269) 
```

```python
# 270. Merge two summaries using state as the key.

#kyuki question mai do summury nikalne  ke liye bol atha toh maine ek nikal state ka total and ek nikala  ir ek nikala state ka average 

state_total =   df.groupby('state').agg({'Marks':'sum'}).reset_index()  # total  hum sum se nikalta  hu 

state_average = df.groupby('state').agg({'Marks':'mean'}) # average mai mean se nikalta hu brother 

df_270 = state_total.merge(state_average, on = 'state' , how = 'inner')

print(df_270)
```

```python
# 271. Merge two summaries using city as the key.


city_total =   df.groupby('city').agg({'Marks':'sum'})
city_average = df.groupby('city').agg({'Marks':'mean'})

df_271 = city_total.merge(city_average , on = 'city' , how = 'inner')

print(df_271)
```

```python
# 272. Create a centre-to-city mapping DataFrame and join it to student data.

centre_to_city = df[['center_name','city']].drop_duplicates()

df_272 = df.merge(centre_to_city , on = ['center_name','city'] , how = 'inner')

print(df_272)
```

```python
# 273. Identify records that do not match during a merge using an indicator column.

# Q273. Merge karte waqt un records ko identify karo jo match nahi hue.

# quesiton mai koi column nahi diya hai  mai randumliye hi le raha hu 

state_total =   df.groupby('state').agg({'Marks' : 'sum'})
state_average = df.groupby('state').agg({'Marks' : 'mean'})

df_273 = state_total.merge(state_average , on = 'state' , how = 'outer')    # outer join isliye lagaya kyuki inner tab lagate hai jab matching value chiaye hoti hai do dataframes ya do cloumn jinko merge kiya uske bich ke  lekin qyestion mai likh ahai ki machng value nahi chiye matlbh  sara data hi chiye hoga  isliy maine yaha pr outer join lagaya '
print(df_273)
```

```python
# 274. Perform a left merge between student data and state statistics.

state_statistics = df.groupby('state').agg({'Marks' : ['sum','max','min','count','std']})

state_average    = df.groupby('state').agg({'Marks' : 'mean'})

df_774 = state_statistics.merge(state_average , on = 'state' , how = 'left')

print(df_774)  
```

```python
# 274. Perform a left merge between student data and state statistics.

# pehel toh question mai hi bola hai ki state  ka merge krna hai  and left join lagana hai 

# Har state ke students ke Marks ka average nikalna
state_statistics = df.groupby('state').agg({'Marks': 'mean'}).reset_index()  # Same state ke students ke Marks ko group karke unka average nikal rahe hain

# State ka average har student ke data ke saath jodna
df_274 = df.merge(state_statistics, on='state', how='left')  # Student data ki saari rows rakhenge aur har student ke state ka average Marks jodenge

print(df_274)
```

```python
# 275. Perform an inner merge between student data and city statistics.


city_statistics = df.groupby('city').agg({'Marks': 'count'}).reset_index()  # Same city ke students ko group karke unke Marks ki total ginti nikal rahe hain

df_275 = df.merge(city_statistics, on='city', how='inner')  # Dono DataFrames mein matching city ke records ko jodkar har student ke saath uski city ka total count add kar rahe hain

print(df_275)
```

```python
# 276. Perform an outer merge between two summary DataFrames.

# yaha hamhe question mai hi bol diya hai ki outer merge lagana hai  kisis bhi do summury data fram mai maine state_total and state_average
state_total   = df.groupby('state').agg({'Marks':'sum'}).reset_index()
state_average = df.groupby('state').agg({'Marks':'sum'}).reset_index()

df_276 = state_total.merge(state_average , on = 'state' , how = 'outer')   # outer join pure data ko deta hai dono tables ya datafram ke  

print(df_276)
```

```python
# 277. Use merge validation to check whether a state-level merge is one-to-many.

# Har state ke students ke total Marks ka sum nikalna
state_toatl = df.groupby('state').agg({'Marks':'sum'}).reset_index()  # Same state ke students ke Marks ko group karke unka total nikal rahe hain aur reset_index() se state ko normal column bana rahe hain

# Original student data ke saath state ka total Marks jodna
df_277 = df.merge(state_toatl, on='state', how='left')  # Original DataFrame ki saari rows preserve hongi aur har student ke state ka matching total Marks judega

print(df_277)
```

```python
# 278. Merge state and city summaries to create a combined location report.

state_total = df.groupby('state').agg({'Marks': 'sum'}).reset_index()  # Har state ke sabhi students ke Marks ka total nikal rahe hain

city_total = df.groupby(['state', 'city']).agg({'Marks': 'sum'}).reset_index()  # State aur city dono ke hisaab se group karke har city ka total Marks nikal rahe hain

df_278 = city_total.merge(state_total, on='state', how='inner')  # Dono DataFrames mein matching data ko merge kiya hu

print(df_278)
```

```python
# 279. Compare two summary DataFrames and identify states with changed averages.

state_summary_1 = df.groupby('state').agg({'Marks': 'mean'}).reset_index() 

state_summary_2 = state_summary_1.groupby('state').agg({'Marks': 'mean'}).reset_index()  # Pehli summary ke average ka dobara state-wise average nikal rahe hain

df_279 = state_summary_1.merge(state_summary_2, on='state', how='inner')  # Dono summaries ko matching state ke basis par merge kar rahe hain

df_279 = df_279[df_279['Marks_x'] != df_279['Marks_y']]  # Dono averages compare karke sirf alag average wali states nikal rahe hain

print(df_279)  
```

```python
# 280. Create a table containing each state, its average Marks, and national average.

state_average = df.groupby('state').agg({'Marks': 'mean'}).reset_index()  # Har state ke students ke Marks ka average nikal rahe hain

national_average = df['Marks'].mean()  # Sabhi students ke Marks ka overall national average nikal rahe hain

state_average['national_average'] = national_average  # Har state ke saamne national average ki value add kar rahe hain

df_280 = state_average  # State-wise average aur national average ko ek table mein rakh rahe hain

print(df_280)  
```

```python
# | Requirement                         | Join    |
# | ----------------------------------- | ------- |
# | Original DataFrame ke saare records | `left`  |
# | Sirf matching records               | `inner` |
# | Summary ke saare records            | `right` |
# | Dono taraf ke saare records         | `outer` |
```

📌 Pandas Merge – Kab Kya Use Karein?
Join	Kab Choose Karein?	Kya Hoga?
left	Jab left DataFrame ka saara data chahiye	Left ka saara data + Right ka matching data
right	Jab right DataFrame ka saara data chahiye	Right ka saara data + Left ka matching data
inner	Jab sirf matching records chahiye	Dono ka sirf matching data
outer	Jab dono DataFrames ka poora data chahiye	Dono ka saara data, match ho ya na ho
