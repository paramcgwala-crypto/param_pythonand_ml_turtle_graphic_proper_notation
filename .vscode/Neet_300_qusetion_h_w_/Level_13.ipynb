```python
import pandas as pd 
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 241. Create a pivot table showing average Marks by state and city.

df_241 = df.groupby(['state','city']).agg({'Marks': 'mean'}).reset_index()

print(df_241)
```

```python
# 242. Create a pivot table showing student count by state and city.

df_242 = df.groupby(['state','city']).agg({'Marks':'count'}).reset_index()
print(df_242)
```

```python
# 243. Create a pivot table showing total Marks by state and city.

df_243 = df.groupby(['state','city']).agg({'Marks':'mean'}).reset_index()
print(df_243)
```

```python
# 244. Create a pivot table showing maximum Marks by state and city.


df_244 = df.groupby(['state','city']).agg({'Marks':'max'}).reset_index()
print(df_244)
```

```python
# 245. Create a pivot table showing minimum Marks by state and city.

df_244 = df.groupby(['state','city']).agg({'Marks':'min'}).reset_index()
print(df_244)
```

```python
# 246. Create a pivot table showing average Marks by state and centre.

df_246 = df.groupby(['state','center_name']).agg({'Marks':'mean'}).reset_index()
print(df_246)
```

```python
# 248. Create a pivot table of Score_Band against state.
df['High_Medium_Low'] = ""
df_248 = df.groupby(['High_Medium_Low','state']).agg({'Marks':'count'})
print(df_248)
```

```python
# 249. Create a crosstab of state and score categories.

df_249 = df.groupby(['High_Medium_Low','state']).agg({'Marks':'count'})
print(df_249)
```

```python
# 250. Create a crosstab of city and Pass_Fail.

city_avg = df.groupby('city').agg({'Marks': 'mean'}).reset_index()

city_avg = city_avg.rename(columns={'Marks': 'city_avg'})

df = df.merge(city_avg, on='city', how='inner')

df['P_F'] = ""

df.loc[df['Marks'] > df['city_avg'], 'P_F'] = 'Pass'

df.loc[df['Marks'] <= df['city_avg'], 'P_F'] = 'Fail'

print(df.groupby(['city', 'P_F']).agg({'Marks': 'count'}).reset_index())
```

```python
# 251. Create a crosstab of state and Pass_Fail.

state_avg = df.groupby('state').agg({'Marks': 'mean'}).reset_index()

state_avg = state_avg.rename(columns={'Marks': 'state_avg'})

df = df.merge(state_avg, on='state', how='inner')

df['P_F'] = ""

df.loc[df['Marks'] > df['state_avg'], 'P_F'] = 'Pass'

df.loc[df['Marks'] <= df['state_avg'], 'P_F'] = 'Fail'

print(df.groupby(['state', 'P_F']).agg({'Marks': 'count'}).reset_index())
```

```python
# 252. Create a crosstab of state and High_Medium_Low.

state_avg = df.groupby('state').agg({'Marks': 'mean'}).reset_index()  # State-wise average Marks nikalna

q25 = df['Marks'].quantile(0.25)  # 25th percentile (Q1) nikalna

q50 = df['Marks'].quantile(0.50)  # 50th percentile (Median) nikalna

q75 = df['Marks'].quantile(0.75)  # 75th percentile (Q3) nikalna

df['High_Medium_Low'] = ""  # Naya column banana categories store karne ke liye

df.loc[df['Marks'] <= q25, 'High_Medium_Low'] = 'Low'  # Q25 se kam ya barabar Marks ko Low dena

df.loc[(df['Marks'] > q25) & (df['Marks'] <= q75), 'High_Medium_Low'] = 'Medium'  # Q25 se Q75 tak Medium dena

df.loc[df['Marks'] > q75, 'High_Medium_Low'] = 'High'  # Q75 se zyada Marks ko High dena

df_252 = df.groupby(['state', 'High_Medium_Low']).agg({'Marks': 'count'}).reset_index()  # State aur category-wise count nikalna

print(df_252)  
```

```python
# 255. Add margins/totals to a suitable pivot table.

df.groupby(['state','city']).agg({'Marks':'sum'}).reset_index()
```

```python
# 256. Normalize a state-by-score-band crosstab to percentages by row.

q25 = df['Marks'].quantile(0.25)  # 25th percentile nikalna

q75 = df['Marks'].quantile(0.75)  # 75th percentile nikalna

df['High_Medium_Low'] = ''  # Naya column create karna

df.loc[df['Marks'] <= q25, 'High_Medium_Low'] = 'Low'  # Low category dena

df.loc[(df['Marks'] > q25) & (df['Marks'] <= q75), 'High_Medium_Low'] = 'Medium'  # Medium category dena

df.loc[df['Marks'] > q75, 'High_Medium_Low'] = 'High'  # High category dena

df_256 = df.groupby(['state', 'High_Medium_Low']).agg({'Marks': 'count'}).reset_index()  # State aur category-wise count

state_total = df.groupby('state').agg({'Marks': 'count'}).reset_index()  # State-wise total count

df_256 = df_256.merge(state_total, on='state', how='inner')  # State total ko merge karna

df_256['Percentage'] = (df_256['Marks_x'] / df_256['Marks_y']) * 100  # Percentage nikalna

print(df_256)  
```

```python
# 257. Normalize a state-by-score-band crosstab to percentages by column.

q25 = df['Marks'].quantile(0.25)  # Marks ka 25th percentile nikalna

q75 = df['Marks'].quantile(0.75)  # Marks ka 75th percentile nikalna

df['High_Medium_Low'] = ''  # Naya column banana jisme score categories store hongi

df.loc[df['Marks'] <= q25, 'High_Medium_Low'] = 'Low'  # Q25 se kam ya barabar Marks ko Low category dena

df.loc[(df['Marks'] > q25) & (df['Marks'] <= q75), 'High_Medium_Low'] = 'Medium'  # Q25 se Q75 tak Marks ko Medium category dena

df.loc[df['Marks'] > q75, 'High_Medium_Low'] = 'High'  # Q75 se zyada Marks ko High category dena

df_257 = df.groupby(['state', 'High_Medium_Low']).agg({'Marks': 'count'}).reset_index()  # State aur score category ke hisaab se count nikalna

category_total = df.groupby('High_Medium_Low').agg({'Marks': 'count'}).reset_index()  # Har score category ka total count nikalna

df_257 = df_257.merge(category_total, on='High_Medium_Low', how='inner')  # Category ke total count ko merge karna

df_257['Percentage'] = (df_257['Marks_x'] / df_257['Marks_y']) * 100  # Column-wise percentage calculate karna

print(df_257)  
```
