```python
import numpy as np
```

```python
# Function
def f(x):
    return (x[0] - 2)**2 + 2*(x[1] + 1)**2 + 3*(x[2] - 3)**2 + 4*(x[3] - 4)**2

# Gradient (derivatives)
def gradient(x):
    return np.array([2*(x[0] - 2),4*(x[1] + 1),6*(x[2] - 3),8*(x[3] - 4)])

# Initial values: all four variables are zero
x = np.array([0.0, 0.0, 0.0, 0.0])

alpha = 0.1
tolerance = 1e-8
max_iter = 1000

# Gradient Descent
for i in range(max_iter):
    grad = gradient(x)

    if np.linalg.norm(grad) < tolerance:
        break

    x = x - alpha * grad

# Final result
print("Iterations:", i)
print("x1 =", x[0])
print("x2 =", x[1])
print("x3 =", x[2])
print("x4 =", x[3])
print("Minimum f(x) =", f(x))
```
