Water Pollution Dataset Analysis 

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel("water_pollution_dataset.xlsx")
```

```python
df.columns
```

```python
df['Country']
```

```python
# Q1. For each city, determine the typical acidity level of the water.Which city has the most acidic water on average?                   # acidic level ka koi bhi column nahi hai toh  mai ph value ka sue kiya column ka 

avg_ph = df.groupby('City').agg({'pH Level': 'mean'})   # City ke according data group karke har city ka average pH nikal rahe hain

print(avg_ph)   # Har city ka average pH screen par dekh rahe hain

lowest = avg_ph['pH Level'].min()   # Sabhi cities ke average pH mein se sabse lowest pH value nikal rahe hain

print(avg_ph[avg_ph['pH Level'] == lowest])   # Jis city ka average pH sabse lowest hai, us city ko print kar rahe hain
```

```python
# 2 In the USA, examine the overall dissolved oxygen levels in water sources. What are the typical values across all records?

oxygen = df.loc[df['Country'] == 'USA']['Dissolved Oxygen (mg/L)'].mean()  # USA ke records ko filter karke Dissolved Oxygen ka average nikal raha hu

print(oxygen)
```

```python
df['City']
```

```python
# 3 Focus on data from Brazil. Which city experiences the highest pollution levels over the 5-year period?

# Brazil ke records ko filter karke City aur Pollution Index columns select kar rahe hain
df_3 = df.loc[df['Country'] == 'Brazil'][['City', 'Pollution Index']]

# Pollution Index ko highest se lowest order mein arrange kar rahe hain
df_3 = df_3.sort_values(by='Pollution Index', ascending=False).head(5)

# Sabse highest pollution wali city ka result print kar rahe hain
print(df_3)
```

```python
df['City']
```

```python
# 4 In Mumbai, identify the date when water pollution was at its worst.

# Filter Mumbai, sort pollution from highest to lowest,
# select the first row, and return its date.

df.loc[df['City'] == 'Mumbai'].sort_values(by='Pollution Index', ascending=False).iloc[0]['Date']
```

```python
# 5 For each country, determine how many pollution records were collected. Which country has the most entries?

df.groupby('Country').agg({'Pollution Index' : 'count'}).sort_values(by = 'Pollution Index', ascending = False)

# Har country ke pollution records count kara hu 
# Pollution Index ke count ko highest se lowest sort kara hu 
# Sabse zyada entries wali country sabse upar aayegi
```

```python
# 6 Compare the average water quality for rivers versus lakes in terms of biological oxygen demand. Which type of water source has more organic matter?

# Sirf River aur Lake ka data select karo
# Water Source ke basis par groups banao
# Har group ka average Biological Oxygen Demand (BOD) nikaalo

df[df['Water Source'].isin(['River', 'Lake'])].groupby('Water Source').agg({'Biological Oxygen Demand (BOD)': 'mean'})   # isin ka matlb  agar wateersourse mai hai river or alke tab unka group kr do 
```

```python
# 7 Examine how pollution levels have changed in different countries over the years. Which country has seen the largest increase in pollution?

# Date ko purane se naye order mein sort karo
# Har country ka first aur last Pollution Index nikalo
# Pollution increase calculate karke highest increase find karo


df_7 = df.sort_values('Date').groupby('Country').agg(  # Date sort karke country-wise group banao
    first=('Pollution Index', 'first'),  # Har country ka pehla Pollution Index nikalo
    last=('Pollution Index', 'last')     # Har country ka aakhri Pollution Index nikalo
)

df_7['Increase'] = df_7['last'] - df_7['first']  # Last aur first ka difference nikalo

df_7.sort_values('Increase', ascending=False).head(1)  # Sabse zyada increase wali country dikhao

```

```python
df.columns
```

```python
# 8 Investigate nitrate levels across various water sources in different countries. How do they compare?

df_8 = df.groupby(['Country', 'Water Source']).agg({'Nitrates (mg/L)': 'mean'}).sort_values('Nitrates (mg/L)', ascending=False)  # Country + source ka average nitrate, highest upar

print(df_8)  
```

```python
df.columns
```

```python
# 9 For New York, track how water pollution has changed over time. Is there a noticeable trend over the 5-year period?

df.loc[df['City'] == 'New York'].groupby('Date').agg({'Pollution Index': 'mean'}).reset_index() # New York ka date-wise average pollution trend
```

```python
# 10 Look at how pollution levels change throughout the year in each city. Which cities experience the most significant seasonal variations?

df['Month'] = pd.to_datetime(df['Date']).dt.month  # Date se Month nikala

df.groupby(['City', 'Month']).agg({'Pollution Index': 'mean'}).reset_index()  # Har city ke har month ka average pollution
```

```python
df.columns
```

```python
# 11 Examine the relationship between pollution and various water quality parameters like dissolved oxygen, biological oxygen demand, and nitrates. Which parameter seems to be most closely linked with pollution levels?

df_11 = df.groupby('Pollution Index').agg({'Dissolved Oxygen (mg/L)' : 'mean','Biological Oxygen Demand (BOD)' : 'mean','Nitrates (mg/L)' : 'mean'})  # Pollution Index ke according teeno water quality parameters ka mean

print(df_11)
```

```python
# 12 Track pollution trends over time for multiple cities. Which cities have improved or worsened over the 5 years?

df['Year'] = pd.to_datetime(df['Date']).dt.year  # Date se Year nikala

df_12 = df.groupby(['City', 'Year']).agg({'Pollution Index': 'mean'}).reset_index()  # Har city aur year ka average pollution nikala

print(df_12)  
```

```python
df.columns
```

```python
# 13 Identify any unusual or extreme values for the pH level in each city. On which days did water acidity or alkalinity differ the most from typical values?

city_avg = df.groupby('City').agg({'pH Level': 'mean'}).reset_index()

# City ke basis par INNER JOIN
df = df.merge(city_avg, on='City', how='inner', suffixes=('', '_avg'))

# Actual pH aur average pH ka difference
df['Difference'] = (df['pH Level'] - df['pH Level_avg']) ** 2

# Sabse bada difference upar
df = df.sort_values('Difference', ascending=False)

# Har city ka extreme record
result = df.groupby('City').head(1)

print(result[['City', 'Date', 'pH Level', 'pH Level_avg', 'Difference']])
```

```python
df.columns
```

```python
# 14 For each country and water source, find out which combinations have the worst pollution levels.

df.groupby([ 'Country','Water Source']).agg({'Pollution Index': 'mean'}).sort_values(by = 'Pollution Index',ascending= False).head(1)  # Country + Water Source group → Max pollution → Sort → Worst combination
```

```python
df.columns
```

```python
# 15 Investigate how heavy metal contamination (Lead and Mercury) has changed over the 5-year period. Are any cities seeing an increase or decrease in contamination?

# City aur Year ke hisaab se Lead aur Mercury ka average
df_15 = df.groupby(['City','Year']).agg({'Lead (mg/L)': 'mean','Mercury (mg/L)': 'mean'}).reset_index()

# City aur Year ko order mein laga rahe hain
df_15 = df_15.sort_values(by=['City','Year'])

# Har city ki first aur last year ki value nikal rahe hain
df_15 = df_15.groupby('City').agg({'Lead (mg/L)': ['first','last'],'Mercury (mg/L)': ['first','last']})

print(df_15)
```

```python
df.columns
```

```python
# 16 Assign each record to a pollution severity category based on overall pollution levels. How do cities compare in terms of the proportion of records in the highest severity category?

df['Severity'] = np.where(df['Pollution Index'] > 70, 'High', np.where(df['Pollution Index'] > 40, 'Medium', 'Low'))  # Pollution ko High, Medium, Low mein divide kiya

df_16 = df.groupby(['City', 'Severity']).agg({'Pollution Index': 'mean'}).reset_index()  # Har city mein har severity ke records count kiye

print(df_16) 

# Insight: Beijing aur Mumbai mein pollution severity sabse high hai, jinka average Pollution Index 74.85 aur 74.75 hai.
```

```python
df.columns
```

```python
# 17 Look at the seasonal patterns of pollution for each city. Are there specific times of year when pollution tends to rise or fall?# 17 Look at the seasonal patterns of pollution for each city. Are there specific times of year when pollution tends to rise or fall?

df['Month'] = pd.to_datetime(df['Date']).dt.month  # Date se Month nikala

df_17 = df.groupby(['City','Month']).agg({'Pollution Index':'mean'}).reset_index()  # City aur Month ka average pollution

df_17 = df_17.sort_values(by=['City','Month'], ascending=True)  # City aur Month ko ascending order mein sort kiya

df_17['Rise_Fall'] = np.where(df_17['Pollution Index'] > 50, 'Rise', 'Fall')  # 50 se zyada = Rise, warna Fall

print(df_17) 
```

```python
df.columns
```

```python
# 18 Create a risk index based on a combination of biological oxygen demand, chemical oxygen demand, and nitrates. Which city is most at risk based on this index?

df['Risk_index'] = df['Biological Oxygen Demand (BOD)'] + df['Chemical Oxygen Demand (COD)'] + df['Nitrates (mg/L)']  # BOD + COD + Nitrates ko add karke Risk Index banaya

df_18 = df.groupby('City').agg({'Risk_index': 'mean'})  # Har city ka average Risk Index nikala

print(df_18)  

# Beijing → 44.98  =  sabse highest Risk Index
```

```python
df.columns
```

```python
# 19 Predict future pollution levels for one city based on past data. What could the pollution level be next year?

df['Year'] = pd.to_datetime(df['Date']).dt.year  # Date se Year nikala

df_19 = df[df['City'] == 'Mumbai'].groupby('Year').agg({'Pollution Index': 'mean'}).reset_index()  # Mumbai ka year-wise average pollution

print(df_19)
```
