```python
import pandas as pd

# Load Dataset
df = pd.read_excel("boston.xlsx")
```

```python
# Task 1 : Check Null Values

print(df.isnull().sum())

# Remove Null Values

df = df.dropna()

print("\nAfter Removing Null Values")

print(df.isnull().sum())
```

```python
# Task 2 : Check Duplicate Rows

print("Duplicate Rows :", df.duplicated().sum())

# Remove Duplicate Rows

df = df.drop_duplicates()

print("\nDuplicate Rows After Removal :", df.duplicated().sum())
```

```python
# Task 3 : Check Categorical Columns

categorical_columns = df.select_dtypes(include="object").columns

print(categorical_columns)

# One Hot Encoding (Only if categorical columns exist)

if len(categorical_columns) > 0:
    df = pd.get_dummies(df, columns=categorical_columns, drop_first=True)
    print("One Hot Encoding Applied")
else:
    print("No Categorical Columns Found")
```

```python
# Dataset ke saare column names dekho

print(df.columns)
```

```python
# Task 5 : Multicollinearity Analysis

corr_matrix = df.corr(numeric_only=True)

print(corr_matrix)
```
