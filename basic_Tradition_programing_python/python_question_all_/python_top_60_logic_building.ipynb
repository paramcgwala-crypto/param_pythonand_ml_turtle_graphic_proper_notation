```python
# Q1. Print numbers from 1 to 10.

for i in range(1, 11):
    print(i)
```

```python
# Q2. Print numbers from 10 to 1.

for i in range(10, 0, -1):
    print(i)
```

```python
# Q3. Print even numbers from 1 to 20.

for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
# Q4. Print odd numbers from 1 to 20.

for i in range(1, 21):
    if i % 2 != 0:
        print(i)
```

```python
# Q5. Print multiples of 5 from 5 to 50.

for i in range(5, 51, 5):
    print(i)
```

```python
# Q6. Print numbers divisible by 3 from 1 to 30.

for i in range(1, 31):
    if i % 3 == 0:
        print(i)
```

```python
# Q7. Check whether a number is even or odd.

num = 10

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Q8. Check whether a number is positive or negative.

num = -5

if num >= 0:
    print("Positive")
else:
    print("Negative")
```

```python
# Q9. Find the greater number between two numbers.

a = 20
b = 15

if a > b:
    print(a)
else:
    print(b)
```
