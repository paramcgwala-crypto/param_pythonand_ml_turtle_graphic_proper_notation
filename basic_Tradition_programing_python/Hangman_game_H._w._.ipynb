Hangman game


```python
import turtle
import random

# Random words ki list
words = ["python", "turtle", "computer", "coding", "programming"]

# List mein se ek random word choose karega
word = random.choice(words)

# User ne jo letters guess kiye hain unko store karega
letters = []

# Wrong guesses count karega
wrong = 0


# Turtle ka object banaya
t = turtle.Turtle()

# Turtle ki drawing speed
t.speed(10)

# Line ko thoda mota rakha
t.pensize(3)


# -------- HANGMAN STAND --------

# Bina line draw kiye position change karega
t.penup()
t.goto(-200, -200)
t.pendown()

# Neeche wali line banayega
t.forward(400)

# Pole ki position
t.penup()
t.goto(-100, -200)
t.pendown()

# Pole upar banayega
t.left(90)
t.forward(300)

# Upar wali line banayega
t.right(90)
t.forward(120)

# Rope banayega
t.right(90)
t.forward(40)


# -------- GAME START --------

# 6 wrong guesses tak game chalega
while wrong < 6:

    # Word ko display karne ke liye empty string
    show = ""

    # Word ke har letter ko check karega
    for x in word:

        # Agar letter guess ho chuka hai
        if x in letters:
            show += x + " "

        # Agar letter guess nahi hua
        else:
            show += "_ "

    # Current word show karega
    print("\nWord:", show)

    # Wrong guesses show karega
    print("Wrong:", wrong)


    # Agar underscore nahi bacha
    # Matlab pura word guess ho gaya
    if "_" not in show:
        print("You Won!")
        break


    # User se letter lega
    guess = input("Enter a letter: ").lower()


    # Sirf ek letter allow karega
    if len(guess) != 1:
        print("Enter only one letter!")
        continue


    # Agar letter pehle guess ho chuka hai
    if guess in letters:
        print("Already guessed!")
        continue


    # Guess kiya hua letter list mein add karega
    letters.append(guess)


    # Check karega ki letter word mein hai ya nahi
    if guess in word:

        # Letter sahi hai
        print("Correct!")


    else:

        # Letter galat hai
        print("Wrong!")

        # Wrong count 1 se increase karega
        wrong += 1


        # 1st wrong = Head
        if wrong == 1:

            t.penup()
            t.goto(20, 180)
            t.pendown()

            # Head banayega
            t.circle(30)


        # 2nd wrong = Body
        elif wrong == 2:

            t.penup()
            t.goto(20, 180)

            # Neeche ki direction
            t.setheading(270)

            t.pendown()

            # Body banayega
            t.forward(100)


        # 3rd wrong = Left hand
        elif wrong == 3:

            t.penup()
            t.goto(20, 130)

            # Left side direction
            t.setheading(210)

            t.pendown()

            # Left hand banayega
            t.forward(50)


        # 4th wrong = Right hand
        elif wrong == 4:

            t.penup()
            t.goto(20, 130)

            # Right side direction
            t.setheading(330)

            t.pendown()

            # Right hand banayega
            t.forward(50)


        # 5th wrong = Left leg
        elif wrong == 5:

            t.penup()
            t.goto(20, 80)

            # Left leg ki direction
            t.setheading(240)

            t.pendown()

            # Left leg banayega
            t.forward(60)


        # 6th wrong = Right leg
        elif wrong == 6:

            t.penup()
            t.goto(20, 80)

            # Right leg ki direction
            t.setheading(300)

            t.pendown()

            # Right leg banayega
            t.forward(60)


# Agar 6 wrong guesses ho gaye
if wrong == 6:

    # Game over message
    print("\nGame Over!")

    # Actual word show karega
    print("Word was:", word)


# Turtle window ko open rakhega
turtle.done()
```
