```python

# Q2.Word Frequency
# words = ["apple", "banana", "apple", "orange", "banana", "apple"]

# Question: Count how many times each word appears in the list.

words = ["apple", "banana", "apple", "orange", "banana", "apple"]

count = {}

for i in words:
    if i in count:
        count[i] = count[i] + 1

        

    else:
        count[i]= 1

print(count)




```

```python
# Q6. Password Strength Checker
# password = "Python123"

# Question: Use a for loop to check whether the password contains:
# - At least one uppercase letter
# - At least one lowercase letter
# - At least one digit
# - At least 8 characters

# Print "Strong Password" if all conditions are satisfied.
```

```python
password = "Python123"

uppercase = False
lowercase = False
digit = False

for i in password:

    if i.isupper():
        uppercase = True

    elif i.islower():
        lowercase = True

    elif i.isdigit():
        digit = True

if uppercase and lowercase and digit and len(password) >= 8:
    print("Strong Password")




```
