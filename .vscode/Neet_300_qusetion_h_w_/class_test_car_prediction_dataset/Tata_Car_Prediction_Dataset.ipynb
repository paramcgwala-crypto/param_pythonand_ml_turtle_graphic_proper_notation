Tata car prise pridiction 

```python
import pandas as pd
```

```python
df = pd.read_csv('tata_car_price_prediction_dataset.csv')
```

```python
df.head(5)
```

```python
df.columns
```

```python
# 1. For each model, calculate the number of cars, average resale price, median resale price, average mileage, 
# and average kilometers driven. Sort the result by average resale price in descending order.

df1 = df.groupby('model').agg({'car_id': 'count','resale_price_lakh': ['mean', 'median'],'mileage_kmpl': 'mean','kilometers_driven': 'mean'})
df1 = df1.sort_values(by=('resale_price_lakh', 'mean'),ascending=False)
print(df1)
```

```python
# 2.Create a new column depreciation_lakh using the difference between ex_showroom_price_lakh and resale_price_lakh. 
# Find the 10 cars with the highest depreciation

df_10 = df.groupby('model').agg({'car_id': 'count', 'resale_price_lakh': 'mean', 'mileage_kmpl': 'mean', 'power_bhp': 'mean', 'accident_history': 'count'}).reset_index()  # Har model ki total cars aur average details

df_10['score'] = df_10['mileage_kmpl'] + df_10['power_bhp'] + df_10['resale_price_lakh']  # Mileage, power aur price ko jodkar score nikala

df_10 = df_10.sort_values(by='score', ascending=False)  # Sabse zyada score wala model upar

print(df_10)
```

```python
# 3. For every model, find the car with the highest resale_price_lakh and display its complete record.

df_3 = df.groupby('model').agg({'resale_price_lakh': 'max'}).reset_index()

df_3 = df.merge(df_3, on=['model', 'resale_price_lakh'], how='inner')

print(df_3)
```

```python
# 4. For every model, calculate the average resale price and identify all cars whose resale price is higher than their model's average resale price.

df_4 = df.groupby('model').agg({'resale_price_lakh': 'mean'}).reset_index()

df_4 = df.merge(df_4, on='model', how='inner')

df_4 = df_4.loc[df_4['resale_price_lakh_x'] > df_4['resale_price_lakh_y']]

print(df_4)
```

```python
#5. Create a car_age column using 2026 as the current year. Calculate the average resale price for each car-age group.

current_year = 2026

df['car_age'] = current_year - df['year']

df_5 = df.groupby('car_age').agg({'resale_price_lakh': 'mean'})

print(df_5)
```

```python
# 6. For every combination of model and fuel_type, calculate the number of cars, average resale price, average mileage, 
# and average power. Consider only groups containing at least 10 cars.



df_6 = df.groupby(['model', 'fuel_type']).agg({'car_id': 'count', 'resale_price_lakh': 'mean','mileage_kmpl': 'mean','power_bhp': 'mean'
}).reset_index()

df_6 = df_6.loc[df_6['car_id'] >= 10]

print(df_6)
```

```python
# 7. For every model, rank cars based on resale_price_lakh, where rank 1 represents the most expensive car within that model.
#  Display the top 3 cars from each model.

df_7 = df.sort_values(by='resale_price_lakh', ascending=False)  # sort values top to bottom mai sort krta hai  ascending false yane  desending true 

df_7 = df_7.groupby('model').head(3)                 # har mdoel ka top 3 cars show kr raha hai 

print(df_7)
```

```python
# 8. Divide cars into four groups based on resale_price_lakh using pd.qcut(). Name the groups Budget, Mid-Range, 
# Premium, and Luxury. Calculate the number of cars, average mileage, average power, and average kilometers driven for e
# ach group.
```

```python
# 9. For every model, calculate the average, median, minimum, maximum, and standard deviation of resale_price_lakh. 
# Then calculate the coefficient of variation (std / mean) and identify the models with the highest 
# variation in resale prices.

df_9 = df.groupby('model').agg({'resale_price_lakh': ['mean', 'median', 'min', 'max', 'std']}) # Average, beech ki value, sabse kam, sabse zyada aur variation})
df_9['CV'] = df_9[('resale_price_lakh', 'std')] / df_9[('resale_price_lakh', 'mean')]          # CV = Standard Deviation / Mean (price mein kitna variation hai)
df_9 = df_9.sort_values(by='CV', ascending=False)                                              # Sabse zyada variation wala model upar kyuki sort value top to bottom sort krtra hai
print(df_9)
```

```python
# 10. Create a model-level summary containing model, number of cars, average resale price, 
# average depreciation percentage, average mileage, average power, and accident rate. 
# Then rank the models based on a custom score using at least three of these metrics.

df_10 = df.groupby('model').agg({'car_id': 'count', 'resale_price_lakh': 'mean', 'mileage_kmpl': 'mean', 'power_bhp': 'mean', 'accident_history': 'count'}).reset_index()  # Har model ki total cars aur average details

df_10['score'] = df_10['mileage_kmpl'] + df_10['power_bhp'] + df_10['resale_price_lakh']  # maine yaha new colun score banane  hai  useke liye ye step kiye. ## Mileage, power aur price ko jodkar score nikala

df_10 = df_10.sort_values(by='score', ascending=False)  # Short values top to bootm sort krti hai ## Sabse zyada score wala model upar

print(df_10)
```
