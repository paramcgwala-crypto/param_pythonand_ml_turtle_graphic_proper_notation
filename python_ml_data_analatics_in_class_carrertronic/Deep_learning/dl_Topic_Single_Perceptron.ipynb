```python
import pandas as pd 
```

```python
data = pd.read_csv("Salary_Data (1).csv")
```

```python
w_old = 100
b_old = 11
h = 0.00001
alpha = 0.01

i = 0

while True:
    f_wb_h = sum((data['Salary'] - ((w_old + h) * data['YearsExperience'] + b_old)) ** 2) / len(data)
    f_wb = sum((data['Salary'] - ((w_old) * data['YearsExperience'] + b_old)) ** 2) / len(data)

    df_dw = (f_wb_h - f_wb) / h

    w_new = w_old - alpha * df_dw

    f_wb_bh = sum((data['Salary'] - ((w_old) * data['YearsExperience'] + (b_old + h))) ** 2) / len(data)
    f_wb = sum((data['Salary'] - ((w_old) * data['YearsExperience'] + b_old)) ** 2) / len(data)

    df_db = (f_wb_bh - f_wb) / h

    b_new = b_old - alpha * df_db

    if round(df_dw, 6) == 0 and round(df_db, 6) == 0:
        break
    else:
        i = i + 1
        w_old = w_new
        b_old = b_new

print(round(df_dw, 6), round(df_db, 6))
print(i, w_new, b_new)
```
