```python
a = [(4,5), (5,6), (-1,-3)]
ed = []

for i in a:
    for i1 in a:
        dis = ((i[0]-i1[0])**2 + (i[1]-i1[1])**2)**0.5
        ed.append(dis)

print(max(ed))
```

```python
a = [('Virat','Batsman'), ('Rohit','Batsman'), ('Bumrah','bowler'),
     ('KL','Batsman'), ('Kuldeep','Bowler')]

d = {'batsman': [], 'bowler': []}

for i in a:
    if i[1] == 'Batsman':
        d['batsman'].append(i[0])
    else:
        d['bowler'].append(i[0])

print(d)
```
