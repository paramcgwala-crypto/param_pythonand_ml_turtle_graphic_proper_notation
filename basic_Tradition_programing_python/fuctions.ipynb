```python
def quad(a, b, c):
    x_1 = (-b + (b**2-4*a*c)**0.5)/2*a
    x_2 = (-b - (b**2-4*a*c)**0.5)/2*a
    return (x_1, x_2)
```

```python
def fact(n):
    f = 1
    for i in range(1, n+1):
        f = f * i
    return f

fact(5)
```

```python
def fact(n):  # Factorial ka function banaya
    a = 1  # Multiplication ki starting value
    for i in range(n, 1, -1):  # n se 2 tak ulta loop
        a = a * i  # Har number ko multiply kiya
    return a  # Final factorial return kiya

fact(5)  # 5 ka factorial nikala
```

```python
# WAP to check whether a number is prime or not

def prime(n):
    if n < 2:
        return False  # 2 se chhota prime nahi hota

    for i in range(2, n):
        if n % i == 0:
            return False  # Puri tarah divide ho gaya

    return True  # Koi divisor nahi mila

prime(7)
```

```python
# WAP to reverse a number

def rev(n):
    a = 0
    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao
    return a

rev(9212394109)
```

```python
# WAP to check whether a number is palindrome or not

def pal(n):
    original = n
    a = 0

    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao

    return original == a  # Original aur reverse compare karo

pal(121)
```

```python
# Task: Python mein def function banao jo check kare ki number Even hai ya Odd, aur return use kare.

def even(n):  # Even-Odd check karne ka function
    if n % 2 == 0:  # Check karo number 2 se divide hota hai ya nahi
        return "even"  # Agar remainder 0 hai toh Even return karo
    else:  # Agar condition False hai
        return "odd"  # Toh Odd return karo

even(8)  # Function ko 8 dekar check kiya
```

```python
# Question 2: Square of a Number
# Task: Python mein def function banao jo kisi number ka square calculate karke return kare.

def square(n):  # Square nikalne ka function
    return n * n  # Number ko khud se multiply karke return karo

square(5)  # Output: 25
```

```python
# Question 3: Sum of Digits
# Task: Python mein def function banao jo kisi number ke sabhi digits ko add karke return kare.

def sum_digits(n):  # Digits ka sum nikalne ka function
    a = 0  # Total sum store karne ke liye

    while n > 0:  # Jab tak number 0 se bada hai
        b = n % 10  # Number ka last digit nikalo
        a = a + b  # Last digit ko total sum mein add karo
        n = n // 10  # Number ka last digit hatao

    return a  # Final sum return karo

sum_digits(325)  # 3 + 2 + 5 = 10
```

```python
# chalo Question 4: Count Digits — ab tu khud logic lagana! 

def count_digits(n):  # Number ke digits count karne ka function
    count = 0  # Digits ka count 0 se start kiya

    while n > 0:  # Jab tak number 0 se bada hai
        count = count + 1  # Har digit ke liye count 1 badhao
        n = n // 10  # Number ka last digit hatao

    return count  # Total digits return karo

count_digits(325)  # Output: 3
```

```python
# Question 5: Largest of Two Numbers
# Task: Python mein def function banao jo do numbers mein se bada number return kare.

def largest(a, b):  # Do numbers mein se bada number nikalne ka function

    if a > b:  # Check karo kya a, b se bada hai
        return a  # Agar haan, toh a return karo
    else:  # Agar a bada nahi hai
        return b  # Toh b return karo

print(largest(10, 25))  # Output: 25
print(largest(50, 20))  # Output: 50
```
