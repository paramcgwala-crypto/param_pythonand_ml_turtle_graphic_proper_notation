```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
df.columns
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 40]

plt.plot(x, y)
plt.show()
```

```python
students = ["A", "B", "C", "D"]
marks = [85, 70, 90, 65]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [50, 80, 30, 60]

plt.bar(products, sales)
plt.title("Product Sales")
plt.show()
```

```python
# **Q. Calculate average marks for each state in the NEET dataset and create a bar chart using Pandas. Keep states on the X-axis and average marks on the Y-axis.**
avg_marks = df.groupby('state')['Marks'].mean()
acg_marks = df.sort_values(by = 'Marks',ascending=False).head(10)
avg_marks.plot(kind='bar')
plt.title('State-wise Average NEET Marks')
plt.xlabel('State')
plt.ylabel('Average Marks')
plt.show()
```

```python

# State-wise average marks
avg_marks = df.groupby('state')['Marks'].mean().reset_index()

# Scatter plot using Pandas
avg_marks.plot(kind='scatter', x='state', y='Marks')

plt.title("State-wise Average NEET Marks")
plt.xlabel("States")
plt.ylabel("Average Marks")
plt.show()
```

```python

# Histogram

marks = [45, 50, 55, 60, 62, 65, 70,
         72, 75, 80, 85, 90, 95, 60, 70]

plt.hist(marks, bins=5, edgecolor='black')
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python

subjects = ["Python", "SQL", "Excel", "Power BI"]
hours = [30, 25, 20, 25]

plt.pie(
    hours,
    labels=subjects,
    autopct='%1.1f%%'
)

plt.title("Study Time Distribution")
plt.show()
```

```python

months = [1, 2, 3, 4, 5]

product_a = [10, 20, 15, 25, 30]
product_b = [15, 18, 22, 20, 35]
product_c = [12, 25, 20, 30, 40]
product_d = [8, 15, 25, 28, 45]

plt.plot(months, product_a, label="Product A")
plt.plot(months, product_b, label="Product B")
plt.plot(months, product_c, label="Product C")
plt.plot(months, product_d, label="Product D")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")

plt.legend()
plt.show()
```

```python

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [5, 10, 8, 15, 20]

plt.plot(
    x, y,
    color='red',
    linestyle='--',
    marker='<',
    linewidth=4,
    markersize=8
)

plt.show()
```

```python

# Multiple plots at once

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Line Plot")

plt.subplot(1, 2, 2)
plt.bar(x, y)
plt.title("Bar Chart")

plt.tight_layout()
plt.show()
```

```python

# Multiple plots at once

import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

# First plot
plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line Plot")

# Second plot
plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.title("Bar Chart")

# Third plot
plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title("Scatter Plot")

# Fourth plot
plt.subplot(2, 2, 4)
plt.hist(y)
plt.title("Histogram")

plt.tight_layout()
plt.show()


```
