```python
# 1. Given a list of numbers, use a for loop to print each element.
# numbers = [10, 20, 30, 40, 50]

numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)


# 2. Given a list of names, use a for loop to print each name.
# names = ["Rahul", "Amit", "Priya", "Neha", "Ravi"]

names = ["Rahul", "Amit", "Priya", "Neha", "Ravi"]

for name in names:
    print(name)


# 3. Given a list of numbers, use a for loop to calculate the sum of all nufmbers.
# numbers = [5, 10, 15, 20, 25]

numbers = [5, 10, 15, 20, 25]

total = 0

for number in numbers:
    total = total + number

print("Sum =", total)


# 4. Count the number of elements in a list using a for loop without using len().
# numbers = [10, 20, 30, 40, 50, 60]

numbers = [10, 20, 30, 40, 50, 60]

count = 0

for number in numbers:
    count = count + 1

print("Number of elements =", count)


# 5. Given a list of numbers, print only the even numbers.
# numbers = [12, 7, 18, 5, 21, 30, 14]

numbers = [12, 7, 18, 5, 21, 30, 14]

for number in numbers:
    if number % 2 == 0:
        print(number)


# 6. Given a list of numbers, print only the odd numbers.
# numbers = [11, 24, 35, 42, 57, 60, 73]

numbers = [11, 24, 35, 42, 57, 60, 73]

for number in numbers:
    if number % 2 != 0:
        print(number)


# 7. Find the largest number in a list using a for loop without using max().
# numbers = [23, 56, 12, 89, 34, 67]

numbers = [23, 56, 12, 89, 34, 67]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number =", largest)


# 8. Find the smallest number in a list using a for loop without using min().
# numbers = [45, 12, 78, 23, 9, 56]

numbers = [45, 12, 78, 23, 9, 56]

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest number =", smallest)


# 9. Count how many even and odd numbers are present in a list.
# numbers = [10, 15, 22, 31, 40, 51, 62, 73]

numbers = [10, 15, 22, 31, 40, 51, 62, 73]

even_count = 0
odd_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count = even_count + 1
    else:
        odd_count = odd_count + 1

print("Even numbers =", even_count)
print("Odd numbers =", odd_count)


# 10. Create a new list containing the square of every number using a for loop.
# numbers = [1, 2, 3, 4, 5, 6]

numbers = [1, 2, 3, 4, 5, 6]

square_numbers = []

for number in numbers:
    square_numbers.append(number ** 2)

print("Square list =", square_numbers)


# 11. Create a new list containing only numbers greater than 50.
# numbers = [23, 67, 12, 89, 45, 72, 34, 91]

numbers = [23, 67, 12, 89, 45, 72, 34, 91]

greater_than_50 = []

for number in numbers:
    if number > 50:
        greater_than_50.append(number)

print("Numbers greater than 50 =", greater_than_50)


# 12. Count how many numbers are greater than 50.
# numbers = [23, 67, 12, 89, 45, 72, 34, 91]

numbers = [23, 67, 12, 89, 45, 72, 34, 91]

count = 0

for number in numbers:
    if number > 50:
        count = count + 1

print("Count =", count)


# 13. Create a new list where every number is multiplied by 5.
# numbers = [2, 4, 6, 8, 10]

numbers = [2, 4, 6, 8, 10]

new_list = []

for number in numbers:
    new_list.append(number * 5)

print("New list =", new_list)


# 14. Calculate the average of all numbers in a list using a for loop.
# numbers = [10, 20, 30, 40, 50]

numbers = [10, 20, 30, 40, 50]

total = 0
count = 0

for number in numbers:
    total = total + number
    count = count + 1

average = total / count

print("Average =", average)


# 15. Ask the user for a number and check whether it exists in the list using a for loop.
# Do not use the 'in' operator.
# numbers = [10, 20, 30, 40, 50]

numbers = [10, 20, 30, 40, 50]

user_number = int(input("Enter a number: "))

found = False

for number in numbers:
    if number == user_number:
        found = True
        break

if found:
    print("Number exists in the list.")
else:
    print("Number does not exist in the list.")


# 16. Count how many times the number 5 appears in the list using a for loop.
# Do not use count().
# numbers = [5, 2, 5, 8, 5, 10, 3, 5]

numbers = [5, 2, 5, 8, 5, 10, 3, 5]

count = 0

for number in numbers:
    if number == 5:
        count = count + 1

print("Number 5 appears", count, "times.")


# 17. Separate the positive and negative numbers into two different lists.
# numbers = [10, -5, 8, -12, 0, 7, -3, 15]

numbers = [10, -5, 8, -12, 0, 7, -3, 15]

positive_numbers = []
negative_numbers = []

for number in numbers:
    if number > 0:
        positive_numbers.append(number)
    elif number < 0:
        negative_numbers.append(number)

print("Positive numbers =", positive_numbers)
print("Negative numbers =", negative_numbers)


# 18. Create a new list containing only unique values using a for loop.
# numbers = [10, 20, 10, 30, 20, 40, 30, 50]

numbers = [10, 20, 10, 30, 20, 40, 30, 50]

unique_values = []

for number in numbers:
    if number not in unique_values:
        unique_values.append(number)

print("Unique values =", unique_values)


# 19. Find the second largest number in a list using a for loop.
# Do not use sort() or sorted().
# numbers = [10, 45, 23, 89, 67, 12, 56]

numbers = [10, 45, 23, 89, 67, 12, 56]

largest = numbers[0]
second_largest = numbers[0]

for number in numbers:
    if number > largest:
        second_largest = largest
        largest = number
    elif number > second_largest and number != largest:
        second_largest = number

print("Second largest number =", second_largest)


# 20. Find the common elements between two lists using a for loop.
# list1 = [10, 20, 30, 40, 50]
# list2 = [30, 40, 50, 60, 70]

list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common_elements = []

for number in list1:
    if number in list2:
        common_elements.append(number)

print("Common elements =", common_elements)
```
