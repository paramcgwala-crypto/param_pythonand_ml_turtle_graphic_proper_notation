Level 12: Advanced Filtering & Comparative Analysis

```python
import pandas as pd
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 221. Find students scoring above the overall average Marks.

avg = df['Marks'].mean()
df_221 = df.loc[df['Marks']>avg]
print(avg)
```

```python
# 222. Find students scoring below the overall average Marks.

avg = df['Marks'].mean()
df_222 = df.loc[df['Marks'] < avg]
print(avg)
```

```python
# 223. Find students scoring equal to the maximum Marks.

maximum_Marks = df['Marks'].max()                    # Logic: Find the highest Marks using max(), then filter students scoring equal to it.
df_223 = df.loc[df['Marks']==  maximum_Marks]        # Critical Thinking: 61 students scored the maximum Marks of 720.
print(df_223)
```

```python
# 224. Find students scoring equal to the minimum Marks.

minimum_Marks = df['Marks'].min()
df_224 = df.loc[df['Marks']== minimum_Marks]
print(df_224)
```

```python
# 225. Find states whose average Marks exceed the national average.

# National average: Poore India ka average Marks
national_average = df['Marks'].mean()

# df225 : Har state ka average Marks
df225 = df.groupby('state').agg({'Marks': 'mean'})

# Filter states whose average Marks exceed the national average
df_225 = df225.loc[df225['Marks'] > national_average]

print(df_225)

```

```python
# 226. Find cities whose average Marks exceed the overall average.

overall_average = df['Marks'].mean()
df226 = df.groupby('city').agg({'Marks':'mean'})        # edxceed yane jada yaha hamne check kiya hai ki kitne  students  ke kitne cietes se overall average marksse  unke marks jada aye hai  ye hamne check kiya hai brother
df_226 = df226.loc[df226['Marks']> overall_average]
print(df226)
```

```python
# 227. Find centres whose average Marks exceed the overall average.

overall_average2 = df['Marks'].mean()
df227 = df.groupby('center_num').agg({'Marks':'mean'})
df_227 = df227.loc[df227['Marks'] > overall_average2]
print(df_227)
```

```python
# 231. Find the state with the highest average Marks.


df231 = df.groupby('state').agg({'Marks':'mean'})
df231 = df231.sort_values(by = 'Marks',ascending = False).head(1)
print(df231)

# Logic: Har state ka average Marks nikalo, phir descending order mein sort karke highest average wali state nikalo.
# Critical Thinking: Sabse zyada average Marks wali state find karne ke liye ascending=False aur head(1) use kiya.
```

```python
# 232. Find the state with the lowest average Marks.

df232 = df.groupby('state').agg({'Marks':'mean'})
df232 = df232.sort_values(by = 'Marks',ascending= True).head(1)
print(df232)
```

```python
# 233. Find the city with the highest average Marks.


df233 = df.groupby('city').agg({'Marks':'mean'})
df233 = df233.sort_values(by = 'Marks',ascending= False).head(1)
print(df233)
```

```python
# 234. Find the city with the lowest average Marks.

df234 = df.groupby('city').agg({'Marks':'mean'})
df234 = df234.sort_values(by = 'Marks',ascending= True).head(1)
print(df234)
```

```python
# 235. Find the centre with the highest average Marks.

df235 = df.groupby('center_num').agg({'Marks':'mean'})
df235 = df235.sort_values(by = 'Marks',ascending= False).head(1)
print(df235)
```

```python
# 236. Find the centre with the lowest average Marks.

df236 = df.groupby('center_num').agg({'Marks':'mean'})
df236 = df236.sort_values(by = 'Marks',ascending= True).head(1)
print(df236)
```

```python
# 237. Find the state with the largest number of 650+ scorers.

df_235 = df.loc[df['Marks'] >= 650]
print(df_235.groupby('state').agg({'Marks' : 'count'}).sort_values(by = 'Marks',ascending = False).head())
```

```python
# 238. Find the city with the largest number of 650+ scorers.
df_238 = df.loc[df['Marks'] >= 650]
print(df_238.groupby('city').agg({'Marks':'count'}).sort_values(by = 'Marks',ascending = False).head())
```

```python
# 239. Find the centre with the largest number of 650+ scorers.

df_238 = df.loc[df['Marks'] >= 650]
print(df_238.groupby('center_name').agg({'Marks':'count'}).sort_values(by = 'Marks',ascending = False).head())
```

```python
# 240. Find centres with more than the average number of students.

df_238 = df.groupby('center_name').agg({'Marks':'count'})    # Pehle har centre ke students count karo.
avg = df_238['Marks'].mean()                                 # Phir sabhi centres ke student count ka average nikalo.
df238 = df_238.loc[df_238['Marks'] > avg]                    # Finally, average se zyada students wale centres filter karo.
print(df238)
```
