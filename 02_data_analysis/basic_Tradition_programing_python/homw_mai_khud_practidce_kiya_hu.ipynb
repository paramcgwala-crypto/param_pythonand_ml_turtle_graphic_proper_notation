```python
# Q1. Second Largest

# numbers = [10, 45, 23, 89, 67, 89, 12]

# max() aur sort() use kiye bina second largest unique number find karo.


number = [10, 45, 23, 89, 67, 89, 12]

largest = 0
second_l = 0

for i in number:
     if i > largest:
       second_l = largest
     largest = i

     if  i > second_l and i < largest:
           second_l = i
print(second_l)

```

```python
# Q2. Frequency Count
# numbers = [2, 3, 2, 5, 3, 2, 7, 5, 5, 5]
# Question:

# Har number kitni baar aaya hai, dictionary mein store karo.

# Expected Output:
# {2: 3, 3: 2, 5: 4, 7: 1}
# 🧠 Restrictions:
# Counter use nahi karna
# count() use nahi karna

# numbers = [2, 3, 2, 5, 3, 2, 7, 5, 5, 5]

numbers = [2, 3, 2, 5, 3, 2, 7, 5, 5, 5]

frequency = {}

for i in numbers:
    if i in frequency:
        frequency[i] = frequency[i] + 1
    else:
        frequency[i] = 1

print(frequency)

```

```python
# Q3. Remove Duplicates
# numbers = [10, 20, 10, 30, 20, 40, 30]

# Question: Duplicate numbers hatao, lekin original order same rehna chahiye.

# Expected output:

# [10, 20, 30, 40]
# Restriction:

# set() use nahi karna.

numbers = [10, 20, 10, 30, 20, 40, 30]

unique =[]

for i  in numbers:
    if i not in unique:   # yaha pr  ab udplicate ko show nahi karega jo phele se unique ai ek baar aa gaye toh duplict ek o nahi dekhega 
        unique.append(i)

    else:
        print("no dublicate found")

print("i remeve the dublicates",unique)







```
