```python
def quad(a, b, c):
    x_1 = (-b + (b**2-4*a*c)**0.5)/2*a
    x_2 = (-b - (b**2-4*a*c)**0.5)/2*a
    return (x_1, x_2)
```

```python
def fact(n):
    f = 1
    for i in range(1, n+1):
        f = f * i
    return f

fact(5)
```

```python
def fact(n):  # Factorial ka function banaya
    a = 1  # Multiplication ki starting value
    for i in range(n, 1, -1):  # n se 2 tak ulta loop
        a = a * i  # Har number ko multiply kiya
    return a  # Final factorial return kiya

fact(5)  # 5 ka factorial nikala
```

```python
# WAP to check whether a number is prime or not

def prime(n):
    if n < 2:
        return False  # 2 se chhota prime nahi hota

    for i in range(2, n):
        if n % i == 0:
            return False  # Puri tarah divide ho gaya

    return True  # Koi divisor nahi mila

prime(7)
```

```python
# WAP to reverse a number

def rev(n):
    a = 0
    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao
    return a

rev(9212394109)
```

```python
# WAP to check whether a number is palindrome or not

def pal(n):
    original = n
    a = 0

    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao

    return original == a  # Original aur reverse compare karo

pal(121)
```
