```python
# WAP to solve quadratic equation

def quad(a, b, c=25):
    x_1 = (-b + (b**2 - 4*a*c)**0.5) / (2*a)
    x_2 = (-b - (b**2 - 4*a*c)**0.5) / (2*a)

    return x_1, x_2

result = quad(1, 5)
print(result)
```

```python

def quad(a, b, c):
    if not isinstance(a, (int, float)):
        raise TypeError("a must be an integer or float")

    if not isinstance(b, (int, float)):
        raise TypeError("b must be an integer or float")

    if not isinstance(c, (int, float)):
        raise TypeError("c must be an integer or float")

    x_1 = (-b + (b**2 - 4*a*c)**0.5) / (2*a)
    x_2 = (-b - (b**2 - 4*a*c)**0.5) / (2*a)

    return (x_1, x_2)


result = quad(1, 5, 6)
print(result)
```

```python

def count_vowels(s):
    if not isinstance(s, str):
        raise TypeError("Input must be a string")

    vowels = "aeiouAEIOU"
    count = 0

    for i in s:
        if i in vowels and isinstance(i, str):
            count += 1

    return count


print(count_vowels("Careertronic"))
```

```python

# Question: Do lists ko merge karke ek list return karni hai.
# Example: [1, 2, 3] + [4, 5, 6] -> [1, 2, 3, 4, 5, 6]

def merge_lists(a, b):
    # isinstance() check karta hai ki variable ka data type kya hai.
    # isinstance(a, list) True dega agar 'a' ek list hai.
    # not True ko False aur False ko True bana deta hai.
    # Isliye 'not' ka use karke hum check kar rahe hain
    # ki 'a' list nahi hai ya 'b' list nahi hai.

    if not isinstance(a, list) or not isinstance(b, list):

        # Agar dono mein se koi bhi input list nahi hai,
        # toh TypeError raise karke program ko galat input
        # ke baare mein clear message denge.
        raise TypeError("Both inputs must be lists")

    # '+' operator dono lists ke elements ko jodta hai.
    # return combined list ko function ke bahar bhejta hai.
    return a + b


# Function ko do lists dekar call kar rahe hain.
result = merge_lists([1, 2, 3], [4, 5, 6])

# Final merged list print hogi.
print(result)

# Output: [1, 2, 3, 4, 5, 6]
```

```python

# Factorial using Recursion

def fact(n):
    # Stopping condition: 0 aur 1 ka factorial 1 hota hai
    if n == 1:
        return 1

    # Recursive call: n ko uske previous number ke factorial se multiply karo
    return n * fact(n - 1)


# Function call
print(fact(5))

# Output: 120
```

```python
# Q1. Fibonacci series using recursion
# 1, 1, 2, 3, 5, 8, 13, 21

def fibonacci(n):
    # Stopping condition: 0 ya 1 tak pahunchne par ruk jao
    if n <= 1:
        return n

    # Pichhle do numbers ko add karke next number nikalo
    return fibonacci(n - 1) + fibonacci(n - 2)


# Pehle 8 Fibonacci numbers print karo
for i in range(8):
    print(fibonacci(i), end=" ")

# Output: 0 1 1 2 3 5 8 13
```

```python
# Q2. Find the sum of numbers
#9617422068


def digit_sum(n):
    # Stopping condition: number 0 hone par ruk jao
    if n == 0:
        return 0

    # Last digit aur baaki digits ka sum add karo
    return n % 10 + digit_sum(n // 10)


# Apna number pass karo
print(digit_sum(9617422068))
```

```python
# Q3. Factorial using recursion
# Image mein factorial ke examples dikh rahe hain, jaise fact(4) aur fact(3).

def factorial(n):
    # Stopping condition: 0 aur 1 ka factorial 1 hota hai
    if n == 0 or n == 1:
        return 1

    # Number ko uske previous number ke factorial se multiply karo
    return n * factorial(n - 1)


# 5 ka factorial nikalo
print(factorial(5))

# Output: 120
```
