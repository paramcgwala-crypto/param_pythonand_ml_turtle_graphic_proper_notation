Level - 11 :  Missing Values, Duplicates & Data Quality

```python
import pandas as pd 
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 201. Count missing values in every column.

df.isnull().sum() 
```

```python
# 202. Display rows containing at least one missing value.

# isnull() → har cell ko check karta hai ki missing hai ya nahi
# sum(axis=1) → har row mein total missing values count karta hai
df_202 = df.isnull().sum(axis=1)

# Jis row mein missing values ka count 0 se greater hai,
# us row mein kam se kam 1 missing value hai
df_202 = df.loc[0]

print(df_202)
```

```python
# 203. Display rows where Marks is missing.

df.loc[df['Marks'].isnull()]
```

```python
# 204. Display rows where state is missing.


df.loc[df['state'].isnull()]
```

```python
# 205. Display rows where city is missing.

df.loc[df['city'].isnull()]
```

```python
# 206. Fill missing Marks with the median Marks.

df_206 = df['Marks'].fillna(df['Marks'].median())

print(df_206)

```

```python
# 207. Fill missing Marks with the mean Marks.

df_207 = df['Marks'].fillna(df['Marks'].mean())
print(df_207)
```

```python
# 208. Fill missing categorical values with 'Unknown'.

df_208 = df['city'].fillna('unknown')
print(df_208)
```

```python
# 209. Drop rows where Marks is missing.

df_209 = df.drop(df[df['Marks'].isnull()].index)

print(df_209)
```

```python
# 210. Drop rows where state or city is missing.

df_210_mssing = df.loc[df['state'].isnull() | df['city'].isnull()].index  # State ya city mein missing value wali rows ke index nikalna

df_210 = df.drop(df_210_mssing)  # Missing value wali rows ko DataFrame se remove karna

print(df_210)  # Missing rows remove hone ke baad final DataFrame print karna
```

```python
# 211. Check for duplicated Srlno values.

df_211 = df['Srlno.'].duplicated().sum()

print(df_211)
```

```python
# 212. Display all duplicated Srlno records.

# duplicated() duplicate Srlno values ko identify karta hai.
df_212 = df.loc[df['Srlno.'].duplicated()]

print(df_212)
```

```python
# 213. Remove duplicate rows while keeping the first occurrence.

df_213 = df.drop_duplicates(keep='first')  # Duplicate rows hata kar pehli row rakhna

print(df_213)  
```

```python
# 214. Remove duplicate rows while keeping the last occurrence.

df_214 = df.drop_duplicates(keep='last')  # Duplicate rows hata kar aakhri row rakhna

print(df_214)  
```

```python
# 215. Check whether center_num is unique within center_name.

df_215 = df.groupby('center_name').agg({'center_num': 'nunique'})  # Har center_name ke unique center_num count karna

print(df_215)  
```

```python
# 216. Check whether every row has a valid Marks value between 0 and 720.

df_216 = df.loc[(df['Marks'] >= 0) & (df['Marks'] <= 720)]  # 0 se 720 ke beech ke Marks wali rows select karna

print(df_216)  
```

```python
# 217. Display rows with Marks outside the valid 0–720 range.

df_217 = df.loc[(df['Marks'] < 0) | (df['Marks'] > 720)].reset_index()  # 0 se kam ya 720 se zyada Marks wali rows nikalna yani ek tarike se outside likha yani invalid 

print(df_217)  
```

```python
# 218. Count invalid Marks values outside the range 0–720.

df_218 = df.loc[(df['Marks'] < 0) | (df['Marks'] > 720)].sum()   # Invalid Marks ki total count nikalna

print(df_218)

# Invalid Marks ka total = -9880 ye hi  mai nikal raha tha  
```

```python
# 219. Check for blank strings in state, city, and center_name.

df_219 = df.loc[(df['state']== " ") | (df['city'] == " ") | (df['center_name'] == " ")]        # In columns mein blank values wali rows nikalna hu mai 

print(df_219)
```

```python
# 220. Replace blank strings with missing values.

df_220 = df.replace(' ',None)      # Blank strings ko missing value banana
   
print(df_220)  # Blank strings ko missing values se replace karne ke baad DataFrame print karna
```
