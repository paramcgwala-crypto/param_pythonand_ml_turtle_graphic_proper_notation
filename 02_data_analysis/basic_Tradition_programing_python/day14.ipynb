```python
a = [50,75,44,100,88,99,110,111]

sum   = 0 

for i in a:
    if i % 11 == 0:
        sum += 1 
    
print(sum)
```

```python
a = [50,75,44,100,88,99,110,111]

sum = 0 
for i in range(len(a)):
    if a[i] % 11 == 0:
        sum += a[i]
print(sum)
```

```python
a = [1,2,2,3,4,3,5,5,5]

b = []

for i in a:
    if i not in b:
        b.append(i)

print(b)
```
