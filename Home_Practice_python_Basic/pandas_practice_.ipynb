```python
import pandas as pd 
import  numpy as np
```

```python
df =pd.read_csv('all_neet_marks_2607.csv')
```

```python
# Marks 600 se greater (> 600) wale saare students nikalo.
 
df.loc[df['Marks'] > 600 ]
```

```python
# CHHATTISGARH state ke aur Marks 500 se greater (> 500) wale saare students nikalo.

df.loc[(df[' state'] == 'CHHATTISGARH') & (df['Marks'] > 500) ]
```

```python
df.columns
```

```python
# q3 NEET dataset df mein Marks ke basis par top 10 students nikalo.

df_3 = df.sort_values( by = 'Marks', ascending=False).head(10)

print(df_3)
```

```python
# 4 NEET dataset df mein har state ka average Marks nikalo.

df.groupby(' state').agg({'Marks' : 'mean'})
```

```python
# 5 Har state ke liye 3 cheezein nikalo:
# Students ki total count
# Average Marks
# Maximum Marks

df.groupby('state').agg({'Marks' : ['count','mean','max']})
```

```python
#6 Har center_name ka average Marks nikalo aur identify karo ki sabse zyada average Marks kis center ka hai.

df_6 = df.groupby('center_name').agg({'Marks': 'mean'})

df_6 = df_6.sort_values(by='Marks', ascending=False).head(1)

print(df_6)
```

```python
# 7 Marks column mein missing (NaN) values kitni hain, aur un rows ko identify karo.

# Marks column mein kitni missing (NaN) values hain
print(df['Marks'].isnull().sum())

# Missing Marks wali saari rows identify karo
print(df.loc[df['Marks'].isnull()])
```

```python
# 8 NEET dataset df mein duplicate rows identify karo aur duplicates ko remove karo, lekin har duplicate ki first occurrence ko keep karo.

df.duplicated().sum()
df.drop_duplicates()
df.drop_duplicates(keep='first')    # first duplicate row ko rakhta hai, baaki duplicate hata deta hai
```

```python
# 9 NEET dataset df mein sirf valid Marks wale students nikalo, jahan Marks > 0 aur Marks <= 720 h

df.loc[(df['Marks'] > 0 ) & (df['Marks'] <= 720)]
```

```python
# 10  NEET dataset df mein Marks ke basis par ek naya column Category banao:

# Marks > 500 → High
# Marks >= 300 and Marks <= 500 → Medium
# Marks < 300 → Low

df['High_Medium_Low'] = np.where(df['Marks'] > 500,'High',np.where(df['Marks'] >= 300,'Medium','Low'))

print(df['High_Medium_Low'])
```

```python
# quesiton acche se likhr kr
# 🟢 Q11 — State-wise High/Medium/Low Students

# NEET dataset df mein har state ke andar kitne students High, Medium aur Low category mein hain, ye count karo.

# Category already High_Medium_Low column mein hai:

# High → Marks > 500
# Medium → Marks 300–500
# Low → Marks < 300

df_11 = df.groupby(['state','High_Medium_Low']).agg({'Marks' : 'count'})

print(df_11)
```

```python
# 🟢 Q12 — State Average Marks Comparison

# NEET dataset df mein har state ka average Marks calculate karo aur us average ko original df mein ek naye column State_Avg_Marks ke roop mein add karo.

state_Avg = df.groupby('state').agg({'Marks' : 'mean'}).reset_index()

state_Avg = state_Avg.rename(columns ={'Marks' : 'State_Avg_Marks'})

df = df.merge(state_Avg, on = 'state' , how = 'inner')

print(df[['state','Marks','State_Avg_Marks']])
```

```python
# 🟢 Q13 — Above / Below State Average

# NEET dataset df mein ek naya column Performance banao.

# Marks ko State_Avg_Marks se compare karo:

# Marks > State_Avg_Marks → Above Average
# Marks < State_Avg_Marks → Below Average
# Marks == State_Avg_Marks → Average

df['Performance']  =  np.where(df['Marks'] > df['State_Avg_Marks'],'Above_Average',np.where(df['Marks'] < df['State_Avg_Marks'],'Below_Average','Average'))

print(df['Performance'])
```

```python
# 🟢 Q14 — Centre-wise Contribution

# NEET dataset df mein har center_name ke total candidates ka percentage contribution calculate karo.

# 🎯 Output mein ye 3 cheezein honi chahiye:
# center_name
# Candidates → us centre ke total students
# Contribution → total candidates mein us centre ka percentage


total_candidates = df.groupby('center_name').agg({'Marks' : 'count'}).reset_index()
# Har centre mein kitne candidates hain, ye count kiya

total_candidates = total_candidates.rename(columns= {'Marks' : 'Candidates'})
# Marks column ka naam Candidates kar diya

total = total_candidates['Candidates'].sum()
# Saare centres ke candidates ko add karke total candidates nikale

total_candidates['Contribution'] = (total_candidates['Candidates'] / total) * 100
# Har centre ka total candidates mein percentage nikala

print(total_candidates['Contribution'])
# Har centre ka contribution percentage print kiya
```

```python
# 🟢 Q15 — State-wise Highest Scorer
# Har state ka highest scorer find karo.
# Output mein ye columns hone chahiye:
# state, Marks, city, center_name

# Hint:
# groupby() + max() se maximum Marks identify karo
# phir original df se us student ki complete row nikalo.


df_15 = df.groupby('state').agg({'Marks': 'max'}).reset_index()
# Har state ka maximum Marks nikala

df_15 = df.merge(df_15, on='state', how='inner')
# Maximum Marks ko original df ke saath state ke basis par merge kiya

df_15 = df_15.loc[df_15['Marks_x'] == df_15['Marks_y']]
# Jinke original Marks state ke maximum Marks ke equal hain, sirf unko rakha

print(df_15[['state', 'Marks_x', 'city', 'center_name']])
# State, highest Marks, city aur centre ka output dikhaya
```

```python
# 🟢 Q16 — State-wise Candidate Count
# Har state mein total kitne candidates hain, ye find karo.
# Output mein state aur Candidates columns hone chahiye.
# Candidates ko descending order mein sort karo.
# Top 10 states show karo.

df_16 = df.groupby('state').agg({'Marks' : 'count'}).reset_index()
df_16 = df_16.rename(columns= {'Marks' : 'Candidates'})
df_16 =  df_16.sort_values(by = 'Candidates',ascending=False).head(10)
print(df_16)
```

```python
# 🟢 Q17 — Above Average Students
# Har state ke average Marks se zyada Marks laane wale students find karo.
# Output mein ye columns hone chahiye:
# state, Marks, city, center_name
#
# Hint:
# Q12 mein banaya hua State_Avg_Marks column use kar.
# loc() se un students ko filter kar jinke Marks,
# State_Avg_Marks se greater hain.

df_17 = df.loc[df['Marks'] > df['State_Avg_Marks']]
# Sirf un students ko filter kiya jinke Marks state average se zyada hain

print(df_17[['state', 'Marks', 'city', 'center_name']])
# Required columns ka output dikhaya
```

```python
# 🟢 Q18 — State-wise High Scorers Percentage

# NEET dataset df mein har state ke un candidates ka percentage calculate karo jinke Marks > 500 hain.

# Output columns:

# state

# High_Scorers

# Total_Candidates

# High_Scorer_Percentage

High_Scorers = df.loc[df['Marks'] > 500]
# 500 se zyada Marks wale candidates filter kiye

High_Scorers = High_Scorers.groupby('state').agg({'Marks': 'count'}).reset_index()
# Har state ke high scorers count kiye

High_Scorers = High_Scorers.rename(columns={'Marks': 'High_Scorers'})
# Marks column ka naam High_Scorers rakha

total_candidates = df.groupby('state').agg({'Marks': 'count'}).reset_index()
# Har state ke total candidates count kiye

total_candidates = total_candidates.rename(columns={'Marks': 'Total_Candidates'})
# Marks column ka naam Total_Candidates rakha

df_18 = High_Scorers.merge(total_candidates, on='state', how='inner')
# Dono tables ko state ke basis par merge kiya

df_18['High_Scorer_Percentage'] = (df_18['High_Scorers'] / df_18['Total_Candidates']) * 100
# Har state mein high scorers ka percentage nikala

print(df_18)
# Final output display kiya
```

```python
# 🟢 Q19 — State-wise Marks Category

# NEET dataset df mein har state ke andar:

# High students ka count

# Medium students ka count

# Low students ka count

# nikalna hai.

# Condition: Q10 mein banaya hua High_Medium_Low column use karna.

# Hint: groupby() mein state aur High_Medium_Low dono use honge, phir agg() se count nikalna hai.

# Chal brother, pehla attempt bhej! 💪

df['High_Medium_Low'] = np.where(df['Marks'] > 500, 'High', np.where(df['Marks'] >= 300, 'Medium','Low'))

df_19 = df.groupby(['state','High_Medium_Low'] ).agg({'Marks' : 'count'})

print(df_19)
```

```python
# 🟢 Q20 — State-wise Top Scorer Details

# NEET dataset df mein har state ke top 3 highest-scoring candidates find karo.

# Output mein ye columns hone chahiye:

# state

# Marks

# city

df_20 = df.sort_values(by= 'Marks' ,ascending=False)

df_20 = df.groupby('state').head(3)

print(df_20)
```

```python
# 🟢 Q21 — State-wise Top Scorer
# NEET dataset df mein har state ka highest-scoring candidate find karo.
# Output columns:
# - state
# - Marks
# - city
# - center_name

df_21 = df.groupby(' state').agg({'Marks': 'max'}).reset_index()
# Har state ke highest Marks nikale aur state ko normal column banaya

df_21 = df.merge(df_21, on=' state', how='inner')
# State ke basis par maximum Marks ko original data ke saath merge kiya

df_21 = df_21.loc[df_21['Marks_x'] == df_21['Marks_y']]
# Sirf un candidates ko rakha jinke Marks state ke maximum Marks ke equal hain

print(df_21[[' state', 'Marks_x', 'city', 'center_name']])
# State, highest Marks, city aur centre ka output dikhaya
```

```python
# 🟢 Q22 — State-wise Lowest Scorer
# NEET dataset df mein har state ka lowest-scoring candidate find karo.
#
# Output columns:
# state, Marks, city, center_name
#
# Rule:
# Har state se sirf lowest Marks wale candidate ki row nikalni hai.
# Agar multiple candidates ke lowest Marks same hain, toh sabko include karna.
#
# Hint:
# Q21 ka pattern use kar, bas maximum ki jagah minimum Marks nikalne hain.

df_22 = df.groupby(' state').agg({'Marks': 'min'}).reset_index()
# Har state ke minimum Marks nikale aur state ko normal column banaya

df_22 = df.merge(df_22, on=' state', how='inner')
# Minimum Marks ko original data ke saath merge kiya

df_22 = df_22.loc[df_22['Marks_x'] == df_22['Marks_y']]
# Sirf lowest Marks wale candidates ko filter kiya

print(df_22[[' state', 'Marks_x', 'city', 'center_name']])
# State, lowest Marks, city aur centre display kiya
```

```python
# 🟢 Q23 — State-wise Median
# NEET dataset df mein har state ke candidates ke median Marks find karo.
# Output columns:
# - state
# - Median_Marks

# 🟢 Q23 — State-wise Median

df_23 = df.groupby(' state').agg({'Marks': 'median'}).reset_index()
# Har state ke median Marks nikale aur state ko normal column banaya

df_23 = df_23.rename(columns={'Marks': 'Median_Marks'})
# Marks column ka naam Median_Marks rakha

print(df_23[[' state', 'Median_Marks']])
# State aur uske median Marks display kiye
```

```python
# 🟢 Q24 — State-wise Marks Variation
# NEET dataset df mein har state ke Marks ka standard deviation calculate karo.
# Output columns:
# - state
# - Marks_Std

df_24 = df.groupby(' state').agg({'Marks': 'std'}).reset_index()
# Har state ke Marks ka standard deviation nikala

df_24 = df_24.rename(columns={'Marks' : 'Marks_Std'})
# Marks column ka naam Marks_Std rakha

print(df_24[[' state', 'Marks_Std']])
# State aur uska standard deviation display kiya
```

```python
# Bhai 🔥 Q25 — State-wise Marks Range
# Task: Har state ke highest aur lowest Marks ka difference nikaalna hai.
# - df use karna hai.
# - Har state ke maximum aur minimum Marks nikalo.
# - Ek naya column Marks_Range banao: Maximum Marks - Minimum Marks.
# - Final output mein sirf state, Max_Marks, Min_Marks, Marks_Range dikhana.

df_25 = df.groupby(' state').agg({'Marks': ['min', 'max']}).reset_index()
# Har state ke minimum aur maximum Marks nikale

df_25.columns = ['state', 'Min_Marks', 'Max_Marks']
# Columns ke naam simple kiye

df_25['Marks_Range'] = df_25['Max_Marks'] - df_25['Min_Marks']
# Maximum aur minimum Marks ka difference nikala

print(df_25[['state', 'Max_Marks', 'Min_Marks', 'Marks_Range']])
# Required columns ka final output dikhaya
```
