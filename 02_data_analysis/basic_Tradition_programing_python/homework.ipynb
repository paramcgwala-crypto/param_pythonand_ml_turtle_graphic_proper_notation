```python
# Q1: Count the even numbers between 10 and 50

count = 0

for i in range(10, 51):
    if i % 2 == 0:
        count = count + 1

print("Even numbers count:", count)
```

```python
# Q2: Sum of even numbers between 10 and 50

total = 0

for i in range(10, 51):
    if i % 2 == 0:
        total += i

print("Sum of even numbers:", total)
```

```python
# Q3: Count numbers divisible by 5, 6 and  10 between 60 and 500

count = 0

for i in range(60, 501):
    if i % 5 == 0 and  i % 6 == 0 and i % 10 == 0:
        count += 1

print("Count:", count)
```

```python
# Q7: Make a spider using turtle
import turtle

t = turtle.Turtle()
t.speed(0)

for i in range(8):
    t.forward(100)
    t.backward(100)
    t.right(45)

t.penup()
t.goto(0, -30)
t.pendown()
t.circle(30)
```

```python
# Q8: Make a flower using turtle

import turtle

t = turtle.Turtle()
t.speed(0)

# Draw flower petals
for i in range(36):
    t.circle(50)
    t.right(10)

# Draw stem
t.left(90)
t.forward(150)
```

```python

# Q9: Count prime numbers between 10 and 1000

count = 0

for num in range(10, 1001):
    factors = 0

    for i in range(1, num + 1):
        if num % i == 0:
            factors += 1

    if factors == 2:
        count += 1

print("Number of prime numbers:", count)
```

```python
# Q10: Find the sum of prime numbers between 10 and 1000

total = 0

for num in range(10, 1001):
    factors = 0

    for i in range(1, num + 1):
        if num % i == 0:
            factors += 1

    if factors == 2:
        total += num

print("Sum of prime numbers:", total)


```
