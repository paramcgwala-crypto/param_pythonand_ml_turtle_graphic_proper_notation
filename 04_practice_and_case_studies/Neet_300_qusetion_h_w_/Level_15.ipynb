```python
import pandas as pd 
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 281. Calculate the overall rank of every student using rank().

df['overall rank'] = df['Marks'].rank(ascending=False)  # Har student ke Marks ke basis par overall rank de raha hu or highest Marks wale ko Rank 1 milegi  second number wale ko rank 3 milega example  jaiseki 750  marks wale ko rank 1  phri suse kam wal eko rank 2 phir usse kam wale ko rank 3 or ese hi  rank milta jayega 

print(df[['Marks','overall rank']])  # Sabhi students ki overall rank check karna

# rank() function se hum students ke Marks ke basis par rank calculate kar sakte hain. Maine ascending=False use kiya hai, isliye highest Marks wale ko highest priority yani Rank 1 milti hai. Agar same Marks hain toh default mein unki average rank aa sakti hai."
```

```python
# 282. Calculate dense rank of students based on Marks.

df['Rank_Of_Student'] = df['Marks'].rank(method='dense', ascending=False)  # Marks ke basis par dense rank dena; same Marks ko same rank milegi aur rank skip nahi hogi

print(df[['Marks','Rank_Of_Student']])  # Har student ki dense rank dekhna

# Dense rank mein hum Marks ke basis par rank dete hain.
# Agar do students ke Marks same hain, toh dono ko same rank milti hai.
# Lekin uske baad wali rank skip nahi hoti, next number ki rank milti hai.

# Example:
# 700, 700, 650, 600
# → 1, 1, 2, 3

# ascending=False ki wajah se highest Marks ko Rank 1 milti hai.
# Rank number jitna chhota hoga, student ki position utni upar hogi.

# Rank_Of_Student column mein har student ki Marks ke basis par dense rank hai.
# Output mein har row ke saamne us student ki rank di hui hai.
# rank() ka result Pandas mein float64 type mein aa sakta hai.

# in hindi 

# यह हमारे पास लगभग 23.3 लाख छात्रों (2,333,162 rows) का रैंक डेटा है। इसमें मुख्य रूप से दो कॉलम्स हैं: Marks और Rank_Of_Student। जैसा कि हम आउटपुट में देख सकते हैं, यह रैंक सीधे तौर पर मार्क्स के आधार पर कैलकुलेट की गई है—जिस छात्र के मार्क्स जितने अधिक हैं (जैसे 449 मार्क्स), उसकी रैंक उतनी ही बेहतर (266.0) आई है। बड़े डेटासेट के कारण Pandas केवल टॉप 5 और बॉटम 5 रोज़ का प्रीव्यू दिखा रहा है।"
```

```python
# 283. Calculate competition rank of students based on Marks.

df['competition_rank'] = df['Marks'].rank(method = 'min',ascending = False)

print(df[['Marks','competition_rank']])

# 1. कोड का लॉजिक
# • ascending=False: जिसके सबसे ज़्यादा मार्क्स होंगे, उसे रैंक 1 मिलेगी।
# • method='min': अगर दो बच्चों के बराबर मार्क्स हैं, तो दोनों को सेम मिनिमम रैंक मिलेगी और अगली रैंक स्किप (छोड़) हो जाएगी। (जैसे: 1, 1 के बाद सीधे 3 रैंक आएगी)।
# 2. आउटपुट का मतलब
# • बड़ा डेटा: यह 23.3 लाख बच्चों की रियल कॉम्पिटिशन रैंक दिखा रहा है।
# • उदाहरण: 449 मार्क्स वाले छात्र की ऑल इंडिया रैंक 2,88,253 है, जबकि 14 मार्क्स वाले की रैंक बहुत पीछे 22,97,740 है।
```

```python
# 284. Calculate percentile rank of every student.

df['Percentile_rank'] = df['Marks'].rank(pct = True) * 100 

print(df[['Marks','Percentile_rank']])

# यह रहा इस कोड और आउटपुट का फाइनल लॉजिक:"
# • 1. कोड क्या कर रहा है?
# "सर, इस कोड में rank(pct=True) का उपयोग करके Marks कॉलम के आधार पर Percentile Rank निकाला गया है। pct=True रैंकिंग को 0 और 1 के बीच लाता है, और फिर * 100 करके हमने उसे 0 से 100 के परसेंटाइल स्केल में बदल दिया है।"
# • 2. डेटा कितना बड़ा है?
# "सर, आउटपुट के नीचे 2333162 rows लिखा है। इससे पता चलता है कि यह किसी छोटे स्कूल का नहीं, बल्कि लगभग 23.3 लाख छात्रों का बहुत बड़ा डेटा है (जैसे JEE या NEET जैसी नेशनल परीक्षा का)।"
# • 3. रिजल्ट का क्या मतलब है? (उदाहरण)
# "सर, परसेंटाइल का मतलब है कि कोई छात्र कितने प्रतिशत बच्चों से आगे है। उदाहरण के लिए:
# 	• जिस छात्र के 449 मार्क्स हैं, उसका परसेंटाइल 87.61% है। यानी वह 87.61% छात्रों से आगे है।
# 	• जिसके सिर्फ 14 मार्क्स हैं, उसका परसेंटाइल 1.47% है, यानी वह सिर्फ 1.47% बच्चों से आगे है।"
```

```python
# 285. Calculate rank within each state.         # each ka matalb hota ahi har 

df['State_ka_Rank']  = df.groupby('state')['Marks'].rank(ascending = False)    # ascending  False  maltab descending True hota hai and  har state ke  andr unke marks ke basis pr maine rank diya hai 

print(df[['state','Marks','State_ka_Rank']])

# 1. कोड का लॉजिक
# • groupby('state'): यह पूरे डेटा को राज्यों (States) के हिसाब से अलग-अलग ग्रुप में बांटता है।
# • rank(ascending=False): यह हर राज्य के अंदर छात्रों को उनके मार्क्स के आधार पर स्टेट रैंक (State Rank) देता है, जिसमें सबसे ज़्यादा मार्क्स वाले को रैंक 1 मिलती है।
# 2. आउटपुट का मतलब
# • स्टेट वाइज रैंकिंग: अब ऑल इंडिया रैंक की जगह हर राज्य की अपनी अलग रैंक दिख रही है।
# • उदाहरण: अंडमान निकोबार के जिस छात्र के 449 मार्क्स हैं, उसकी अपने राज्य में 55वीं रैंक है। वहीं वेस्ट बंगाल के 14 मार्क्स वाले छात्र की अपने राज्य में 1,15,514वीं रैंक है।
# सर आपसे पूछ सकते हैं कि आउटपुट में रैंक .5 (जैसे 321.5, 34694.5) में क्यों आ रही है? क्या आप इसका कारण और इसे ठीक करने का तरीका जानना चाहते हैं?
```

```python
# 294. Find the difference between each student's Marks and the highest Marks in their state.

Student_Highest_Marks = df.groupby('state').agg({'Marks': 'max'}).reset_index()  # Har state ka highest Marks nikalna

df1 = df.merge(Student_Highest_Marks, on='state', how='left')  # Har student ke state ka highest Marks lana

df1['Differences'] = df1['Marks_x'] - df1['Marks_y']  # Student ke Marks me se state ke highest Marks minus kiya hu mai  toh unka difference pata chal gaya mereko 

print(df1[['state', 'Marks_x', 'Marks_y', 'Differences']])  # State, student Marks, highest Marks aur difference dekhna
```

```python
# 295. Find the difference between each student's Marks and the average Marks of their state.

Student_Average_Marks = df.groupby('state').agg({'Marks': 'mean'}).reset_index()  # Har state ke students ka average Marks nikalna

df_295 = df.merge(Student_Average_Marks, on='state', how='left')  # Har student ke saamne uske state ka average Marks lana

df_295['Differences_2'] = df_295['Marks_x'] - df_295['Marks_y']  # Student ke Marks aur state ke average Marks ka difference nikalna

print(df_295[['state', 'Marks_x', 'Marks_y', 'Differences_2']])  # State, student Marks, average Marks aur difference dekhna
```

```python
# 296. Find the percentage of the state maximum achieved by each student.

state_maximum = df.groupby('state').agg({'Marks': 'max'}).reset_index()  # Har state ko group karke us state ka sabse highest Marks nikalna

df_296 = df.merge(state_maximum, on='state', how='left')  # Har student ke state ke saamne us state ka highest Marks add karna

df_296['Percentage'] = (df_296['Marks_x'] / df_296['Marks_y']) * 100  # Student ke Marks ko state ke highest Marks se divide karke percentage nikalna

print(df_296[['state', 'Marks_x', 'Marks_y', 'Percentage']])  # State, student ke Marks, state maximum aur calculated percentage show karna
```

```python
# 297. Find the percentage of the centre maximum achieved by each student.

center_maximum = df.groupby('center_name').agg({'Marks': 'max'}).reset_index()  # Har centre ko group karke us centre ka sabse highest Marks nikalna

df_297 = df.merge(center_maximum, on='center_name', how='left')  # Har student ke centre ke saamne us centre ka highest Marks add karna

df_297['Percentage_2'] = (df_297['Marks_x'] / df_297['Marks_y']) * 100  # Student ke Marks ko centre ke highest Marks se divide karke percentage nikalna

print(df_297[['center_name', 'Marks_x', 'Marks_y', 'Percentage_2']])  # Centre, student Marks, centre maximum aur percentage show karna

# Marks_x  → left/original DataFrame ka Marks
# Marks_y  → right/merged DataFrame ka Marks
```

```python
# 298. Calculate each state's contribution to total candidates.

state_count = df.groupby('state').agg({'Marks': 'count'}).reset_index()  # Har state ko group karke us state mein kitne candidates hain unka count nikalna

total_candidates = len(df)  # Puri DataFrame mein total candidates ki sankhya nikalna

state_count['Contribution'] = (state_count['Marks'] / total_candidates) * 100  # State ke candidates ko total candidates se divide karke percentage contribution nikalna

print(state_count[['state', 'Marks', 'Contribution']])  # Har state ke candidates count aur total mein unki percentage contribution show kr raha hu
```

```python
# 299. Calculate each city's contribution to total candidates.

city_total = df.groupby('city').agg({'Marks': 'count'}).reset_index()  # Har city ko group karke us city mein total kitne candidates hain unka count nikalna

total_candidates_of_city = len(df)  # Puri DataFrame mein sabhi cities ke mila kar total candidates ki sankhya nikalna

city_total['Contribution_2'] = (city_total['Marks'] / total_candidates_of_city) * 100  # Har city ke candidates ko total candidates se divide karke percentage contribution nikalna

print(city_total[['city', 'Marks', 'Contribution_2']])  # City ka naam, uske total candidates aur total candidates mein uski percentage contribution show kr raha hu 
```

```python
# 300. Calculate each centre's contribution to total candidates.

center_total_2 = df.groupby('center_name').agg({'Marks': 'count'}).reset_index()  # Har centre ko group karke us centre mein total kitne candidates hain unka count nikalna

total_candidates_in_center = len(df)  # Puri DataFrame mein sabhi centres ke mila kar total candidates ki sankhya nikalna

center_total_2['Contribution'] = (center_total_2['Marks'] / total_candidates_in_center) * 100  # Har centre ke candidates ko total candidates se divide karke percentage contribution nikalna

print(center_total_2[['center_name', 'Marks', 'Contribution']])  # Centre ka naam, candidates count aur total candidates mein uski percentage contribution show karna
```
