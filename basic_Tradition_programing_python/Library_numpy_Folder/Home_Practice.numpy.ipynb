```python
# Sawal 7 (thoda tough): Pehli 2 rows aur pehle 3 columns nikaal. Output aisa aana chahiye:

# [[1 2 3]
#  [2 3 4]]

a[:2 ,:3]                #pele maine 2 rows nikali nad fir maine 3 column nikala 
```

```python
# Sawal 8: Sirf last 2 columns nikaal, saari rows ke saath. Output:

# [[4 5]
#  [5 6]
#  [6 7]]

a[:,-2:]  # saari row : column ka matlb saari rows adn -1 index hotia hai last  ke elemet y elast row ya last colon ko accses kr ne lke liye revrese indexing bhi bolt ehi 
```

```python
# Sawal 9: Beech ka element nikaal, matlab row 1, column 2. Sirf ek number.

a[1,2]
```

```python
# Sawal 10 (tough): Row 1 aur row 2 ke column 1 se column 3 tak nikaal. Output: 
# [[3 4 5]
#  [4 5 6]]

a[1:3,1:4]      # 1:3 -> row 1 aur row 2 aayengi (start shamil hota hai, end wala 3 shamil nahi hota)
# 1:4 -> column 1, 2, 3 aayenge (start shamil hota hai, end wala 4 shamil nahi hota)
# Output -> [[3 4 5],
#            [4 5 6]]
```

```python
# Sawal 11: Sirf row 0 aur row 2 ek saath nikaal. Output:

# [[1 2 3 4 5]
#  [3 4 5 6 7]]

a[[0,2]]
```

```python
# Sawal 12: Har dusra column nikaal (column 0, 2, 4), saari rows ke saath. Output:

# [[1 3 5]
#  [2 4 6]
#  [3 5 7]]

a[: , [0,2,4]]
```

```python
# Sawal 13: Poori array ki rows ka order ulta kar de (last row pehle). Output:

# [[3 4 5 6 7]
#  [2 3 4 5 6]
#  [1 2 3 4 5]]

a[: : -1 ]

# start:end:step format hai, start aur end khali = poori range
# step -1 = ulta chalo (last row se pehli row ki taraf)
# Output -> [[3 4 5 6 7],
#            [2 3 4 5 6],
#            [1 2 3 4 5]]
```
