```python
# q2. Design a code of bollywood game

# 1. गेम सेटअप (फिल्म का नाम और चांस)
movie = np.array(['S', 'H', 'O', 'L', 'A', 'Y'])
hidden = np.array(['_', '_', '_', '_', '_', '_'])
chances = 9  # BOLLYWOOD के 9 अक्षर

# 2. गेम का लूप
while chances > 0 and '_' in hidden:
    print("\nMovie:", hidden)
    guess = input("Enter a letter: ").upper()
    
    # अगर अक्षर फिल्म में है, तो खाली जगह पर भरें
    if guess in movie:
        hidden[movie == guess] = guess
    else:
        chances = chances - 1
        print("Wrong guess! Chances left:", chances)

# 3. फाइनल रिजल्ट
if '_' not in hidden:
    print("🎉 You Won!")
else:
    print("💥 Game Over!")
```

```python
# q3. Design a hangman game using turtle graphics where parts of the hanging man are drawn on every wrong guess

import turtle
import random

# 1. टर्टल स्क्रीन सेटअप (फाँसी का फंदा बनाने के लिए)
screen = turtle.Screen()
screen.setup(600, 600)
t = turtle.Turtle()
t.speed(0)
t.pensize(4)
t.hideturtle()

# फाँसी का स्टैंड (Gallows) बनाना
def draw_gallows():
    t.penup()
    t.goto(-150, -150)
    t.pendown()
    t.forward(100) # बेस
    t.backward(50)
    t.left(90)
    t.forward(250) # खंभा
    t.right(90)
    t.forward(100) # ऊपरी बीम
    t.right(90)
    t.forward(40)  # रस्सी

draw_gallows()

# गलत अनुमान पर एक-एक करके शरीर के हिस्से बनाना
def draw_hangman(wrong_guesses):
    if wrong_guesses == 1: # सिर (Head)
        t.right(90)
        t.circle(20)
        t.left(90)
        t.penup()
        t.forward(40)
        t.pendown()
    elif wrong_guesses == 2: # शरीर (Body)
        t.forward(60)
    elif wrong_guesses == 3: # बायाँ हाथ (Left Arm)
        t.backward(45)
        t.right(45)
        t.forward(35)
        t.backward(35)
        t.left(45)
    elif wrong_guesses == 4: # दायाँ हाथ (Right Arm)
        t.right(135)
        t.forward(35)
        t.backward(35)
        t.left(135)
        t.forward(45)
    elif wrong_guesses == 5: # बाईं टाँग (Left Leg)
        t.right(45)
        t.forward(45)
        t.backward(45)
        t.left(45)
    elif wrong_guesses == 6: # दाईं टाँग (Right Leg)
        t.right(135)
        t.forward(45)

# 2. गेम लॉजिक और वर्ड्स
words = ["PYTHON", "TURTLE", "MATRIX", "NUMPY", "CODING"]
secret_word = random.choice(words)
guessed_word = ["_" for _ in secret_word]
wrong_attempts = 0
guessed_letters = []

# 3. गेम का लूप
while wrong_attempts < 6 and "_" in guessed_word:
    print("\nWord:", " ".join(guessed_word))
    print(f"Chances left: {6 - wrong_attempts}")
    
    # टर्टल के इनपुट बॉक्स से या कंसोल से इनपुट लेना
    guess = screen.textinput("Hangman", "Enter a letter:").upper()
    
    if not guess or len(guess) != 1 or guess in guessed_letters:
        continue
        
    guessed_letters.append(guess)
    
    if guess in secret_word:
        for i in range(len(secret_word)):
            if secret_word[i] == guess:
                guessed_word[i] = guess
    else:
        wrong_attempts += 1
        draw_hangman(wrong_attempts)

# 4. फाइनल रिजल्ट
if "_" not in guessed_word:
    print(f"\n🎉 You Won! The word was: {secret_word}")
    t.penup()
    t.goto(-50, 150)
    t.write("You Won! 🎉", font=("Arial", 20, "bold"))
else:
    print(f"\n💥 Game Over! The word was: {secret_word}")
    t.penup()
    t.goto(-50, 150)
    t.write("Game Over! 💥", font=("Arial", 20, "bold"))

turtle.done()
```
