````python
# Q1. Count Even Numbers
# Mid-level

# Task: count_even(numbers) naam ka function banao jo list mein even integers count karke return kare.
# Conditions:
# 1. Input list honi chahiye, warna TypeError raise karo.
# 2. for loop se har element check karo.
# 3. Har element integer hona chahiye.
# 4. True, False, float aur string allowed nahi hain.
# 5. Agar koi invalid element mile, toh TypeError raise karo.
# 6. Sirf even numbers count karke return karo.

# Q1. Count Even Numbers

# Even numbers count karne ke liye function banana
def Count_Even_Number(numbers):

    # Check karna ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError("Sirf list allowed hai")

    # Counting ko 0 se start karna
    count = 0

    # List ke har element ko check karna
    for i in numbers:

        # Pehle Boolean values ko check karna
        # Kyunki Python mein bool, int ka subclass hai
        if isinstance(i, bool):
            raise TypeError("True aur False allowed nahi hain")

        # Check karna ki element integer hai ya nahi
        if not isinstance(i, int):
            raise TypeError("Sirf integer values allowed hain")

        # Check karna ki number even hai ya nahi
        if i % 2 == 0:
            count = count + 1

    # Total even numbers ki counting return karna
    return count


# Function ko call karke result check karna
print(Count_Even_Number([3, 6, 9, 9, 6, 3, 6, 9, 3]))  # 3
print(Count_Even_Number([1, 3, 5, 7]))                 # 0
print(Count_Even_Number([2, 4, 6, 8]))                 # 4
````

````python
# Q2. Count Positive Numbers
# Mid-level

# Task: Count_Positive(numbers) naam ka function banao jo list mein positive numbers count karke return kare.
# Conditions:
# 1. Input list honi chahiye, warna TypeError raise karo.
# 2. for loop se har element check karo.
# 3. Har element int ya float hona chahiye.
# 4. True aur False allowed nahi hain.
# 5. String ya kisi doosre invalid type par TypeError raise karo.
# 6. Sirf zero se bade numbers count karo. 0 positive nahi hai.

# Positive numbers count karne ke liye function banana
def Count_Positive(numbers):

    # Check karna ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError('Sirf list allowed hai')

    # Counting ko 0 se start karna
    count = 0

    # List ke har element ko check karna
    for i in numbers:

        # Check karna ki element int ya float hai ya nahi
        if not isinstance(i, (float, int)):
            raise TypeError('Sirf float aur int allowed hain')

        # Check karna ki Boolean value hai ya nahi
        if isinstance(i, bool):
            raise TypeError('Boolean values allowed nahi hain')

        # Check karna ki element string hai ya nahi
        if isinstance(i, str):
            raise TypeError('String values allowed nahi hain')

        # Check karna ki number positive hai ya nahi
        if i > 0:
            count = count + 1

    # Total positive numbers ka count return karna
    return count


# Function ko call karke result check karna
print(Count_Positive([30, 60, 90]))  # Output: 3
````

````python
# Q3. Find the Largest Number
# Mid-level

# Task: Find_Largest(numbers) naam ka function banao jo list ka sabse bada number return kare.
# Conditions:
# 1. Input list honi chahiye, warna TypeError raise karo.
# 2. List empty nahi honi chahiye. Empty list par ValueError raise karo.
# 3. Har element int ya float hona chahiye.
# 4. True, False aur strings allowed nahi hain.
# 5. for loop se largest number find karna hai.
# 6. Built-in max() function use nahi karna.

def Find_Largest(numbers):

    # Check karna ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError("Sirf list allowed hai")

    # Check karna ki list empty toh nahi hai
    if len(numbers) == 0:
        raise ValueError("Empty list allowed nahi hai")

    # Har element ko validate karna
    for i in numbers:

        # Pehle Boolean values ko reject karna
        if isinstance(i, bool):
            raise TypeError("True aur False allowed nahi hain")

        # Check karna ki element int ya float hai
        if not isinstance(i, (int, float)):
            raise TypeError("Sirf int aur float allowed hain")

    # Pehle element ko largest maan lena
    largest = numbers[0]

    # Har number ko largest se compare karna
    for i in numbers:

        # Agar current number bada hai toh largest update karna
        if i > largest:
            largest = i

    # Sabse bada number return karna
    return largest


# Function ko test karna
print(Find_Largest([30, 60, 90]))      # 90
print(Find_Largest([-5, -2, -10]))    # -2
print(Find_Largest([2.5, 8.7, 4.1]))  # 8.7
````

````python
# Q4. Count Duplicate Numbers

# Task: Count_Duplicates(numbers) naam ka function banao.

# Conditions:

# Input list honi chahiye, warna TypeError raise karo.

# Empty list hone par ValueError raise karo.

# List ke elements sirf int hone chahiye.

# True, False, float aur string allowed nahi hain.

# for loop aur if condition ka use karo.

def Count_Duplicates(numbers):

    # Check karna ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError("Sirf list allowed hai")

    # Check karna ki list empty toh nahi hai
    if len(numbers) == 0:
        raise ValueError("Empty list allowed nahi hai")

    # List ke har element ka data type check karna
    for i in numbers:

        # Pehle Boolean values ko reject karna
        if isinstance(i, bool):
            raise TypeError("True aur False allowed nahi hain")

        # Check karna ki element integer hai ya nahi
        if not isinstance(i, int):
            raise TypeError("Sirf integers allowed hain")

    # Duplicate numbers ki counting 0 se start karna
    count = 0

    # set() se duplicate values ko ek baar hi lena
    for i in set(numbers):

        # Check karna ki number ek se zyada baar aaya hai
        if numbers.count(i) > 1:

            # Duplicate numbers ki counting badhana
            count = count + 1

    # Total duplicate numbers ki count return karna
    return count


# Function ko test karna
print(Count_Duplicates([2, 5, 2, 8, 5, 5]))  # 2
print(Count_Duplicates([1, 2, 3, 4]))        # 0
print(Count_Duplicates([7, 7, 7, 7]))        # 1
````

````python
# ## Q5. Count Negative Numbers

# **Task:** `Count_Negative(numbers)` naam ka function banao.

# **Conditions:**

# 1. Input `list` honi chahiye, warna `TypeError` raise karo.
# 2. Empty list hone par `ValueError` raise karo.
# 3. List ke elements sirf `int` ya `float` hone chahiye.
# 4. `True`, `False` aur strings allowed nahi hain.
# 5. `for` loop aur `if` condition ka use karo.
# 6. Sirf negative numbers count karke `return` karo.

# **Expected Output:**

# ```python
# print(Count_Negative([10, -5, -2, 8, -1]))  # 3
# print(Count_Negative([1, 2, 3, 4]))         # 0
# print(Count_Negative([-2, -5, -8]))         # 3
# print(Count_Negative([2.5, -1.5, 3.5]))     # 1
# print(Count_Negative([]))                   # ValueError
# print(Count_Negative([2, True, -5]))        # TypeError
# ```


def Count_Negative(numbers):
    # Check karo ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError('Its allowed to use a list')

    # Empty list ko allow nahi karna
    if len(numbers) == 0:
        raise ValueError('Its no allowed to use empty list')

    # List ke har element ka data type check karo
    for i in numbers:
        # Sirf int aur float allow hain
        if not isinstance(i, (int, float)):
            raise TypeError('Its only allowed int and float')

        # Boolean aur string ko allow nahi karna
        if isinstance(i, (bool, str)):
            raise TypeError('Its not allowed bolean and string are not allowed')

    count = 0  # Negative numbers ki counting 0 se start

    # List ke har number ko check karo
    for i in numbers:
        if i < 0:  # Agar number 0 se chhota hai
            count = count + 1  # Count ko 1 se increase karo

    return count  # Total negative numbers return karo


print(Count_Negative([10, -5, -2, 8, -1, -9]))
````

````python
# Q6. Calculate Sum of Positive Numbers

# Task: Sum_Positive(numbers) naam ka function banao.

# Conditions:

# Input list honi chahiye, warna TypeError raise karo.

# Empty list hone par ValueError raise karo.

# List ke elements sirf int ya float hone chahiye.

# True, False aur strings allowed nahi hain.

# for loop aur if condition ka use karo.

# Sirf positive numbers ka sum calculate karke return karo.

# sum() function use nahi karna hai.

def sum_Positive(numbers):
    if not isinstance(numbers,list):
        raise TypeError('Only list are allowed')
    if len(numbers) == 0:
        raise ValueError('Empty list are not alowed')
    for i in numbers:
        if not isinstance(i,(int,float)):
            raise TypeError('List ke elements sirf int ya float hone chahiye')
        if isinstance(i,(bool,str)):
            raise TypeError('True, False aur strings allowed nahi hain')
    total = 0 
    for i in numbers:
        if i > 0:
            total = total + i 
    return total

print(sum_Positive([3,6,9,-5,-10,-7]))
````

````python
## Q7. Find the Smallest Number

# **Task:** `Find_Smallest(numbers)` naam ka function banao.

# **Conditions:**

# 1. Input `list` honi chahiye, warna `TypeError` raise karo.
# 2. Empty list hone par `ValueError` raise karo.
# 3. List ke elements sirf `int` ya `float` hone chahiye.
# 4. `True`, `False` aur strings allowed nahi hain.
# 5. `for` loop aur `if` condition ka use karo.
# 6. List ka sabse chhota number find karke `return` karo.
# 7. `min()` function use nahi karna hai.

# **Expected Output:**

# ```python
# print(Find_Smallest([10, 5, 20, 3, 8]))  # 3
# print(Find_Smallest([-5, -2, -10]))     # -10
# print(Find_Smallest([2.5, 1.5, 3.5]))   # 1.5
# print(Find_Smallest([]))                # ValueError
# print(Find_Smallest([2, True, 6]))      # TypeError
# print(Find_Smallest([2, "5", 6]))       # TypeError
# ```

# **Rules:**
# - Pehle khud code likhna.Find
# - `min()` use nahi karna.
# - Atak jao toh hint maang lena; seedha solution nahi dekhna.


# Q7. Find the Smallest Number

def Find_Smallest(numbers):

    # Check karna ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError("Sirf list allowed hai")

    # Check karna ki list empty toh nahi hai
    if len(numbers) == 0:
        raise ValueError("Empty list allowed nahi hai")

    # Har element ka data type check karna
    for i in numbers:

        # Pehle Boolean values ko reject karna
        if isinstance(i, bool):
            raise TypeError("True aur False allowed nahi hain")

        # Sirf int aur float allow karna
        if not isinstance(i, (int, float)):
            raise TypeError("Sirf int aur float allowed hain")

    # Pehle element ko smallest maan lena
    smallest = numbers[0]

    # Har number ko smallest se compare karna
    for i in numbers:

        # Agar current number chhota hai toh smallest update karna
        if i < smallest:
            smallest = i

    # Sabse chhota number return karna
    return smallest


print(Find_Smallest([10, 5, 20, 3, 8]))  # 3
print(Find_Smallest([-5, -2, -10]))      # -10
print(Find_Smallest([2.5, 1.5, 3.5]))    # 1.5
````

````python
# ## Q8. Count Numbers Greater Than 10

# **Task:** `Count_Greater_Than_10(numbers)` naam ka function banao.

# **Conditions:**

# 1. Input `list` honi chahiye, warna `TypeError` raise karo.
# 2. Empty list hone par `ValueError` raise karo.
# 3. List ke elements sirf `int` ya `float` hone chahiye.
# 4. `True`, `False` aur strings allowed nahi hain.
# 5. `for` loop aur `if` condition ka use karo.
# 6. Sirf un numbers ko count karo jo `10` se **strictly greater** hain.
# 7. `sum()`, `max()` aur `min()` use nahi karna.

# **Expected Output:**

# ```python
# print(Count_Greater_Than_10([5, 12, 20, 8, 15]))  # 3
# print(Count_Greater_Than_10([1, 2, 3, 10]))       # 0
# print(Count_Greater_Than_10([11, 10.5, 9]))       # 2
# print(Count_Greater_Than_10([]))                  # ValueError
# print(Count_Greater_Than_10([12, True, 20]))      # TypeError
# print(Count_Greater_Than_10([12, "15", 20]))      # TypeError
# ```

# **Practice rules:**
# - Pehle khud code likhna.
# - Abhi solution nahi dekhna.
# - Atak jao toh hint maang lena.


def Count_Greater_Than_10(numbers):

    # Input list check karna
    if not isinstance(numbers, list):
        raise TypeError("Sirf list allowed hai")

    # Empty list check karna
    if len(numbers) == 0:
        raise ValueError("Empty list allowed nahi hai")

    # Har element ka data type check karna
    for i in numbers:
        if isinstance(i, bool):
            raise TypeError("True aur False allowed nahi hain")

        if not isinstance(i, (int, float)):
            raise TypeError("Sirf int aur float allowed hain")

    # Counting 0 se start karna
    count = 0

    # 10 se bade numbers count karna
    for i in numbers:
        if i > 10:
            count = count + 1

    # Total count return karna
    return count


print(Count_Greater_Than_10([5, 12, 20, 8, 15]))  # 3
````
