Level 9: New Columns & Feature Engineering

```python
import pandas as pd
```

```python
df =pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 161 Create a column Pass_Fail where students scoring at least 300 are Pass, otherwise Fail.

df['Pass_Fail'] = 'Fail'  # Pehle sabhi students ko Fail maan lena

df.loc[df['Marks'] >= 300, 'Pass_Fail'] = 'Pass'  # 300 ya usse zyada Marks walon ko Pass karna

print(df)  # Updated DataFrame check karna
```

```python
# 162. Marks ke basis par Grade banana

df['Grade'] = ''  # Naya Grade column banaya

df.loc[df['Marks'] >= 80, 'Grade'] = 'A'  # 80 ya usse zyada = A

df.loc[(df['Marks'] >= 60) & (df['Marks'] < 80), 'Grade'] = 'B'  # 60 se 79 = B

df.loc[(df['Marks'] >= 40) & (df['Marks'] < 60), 'Grade'] = 'C'  # 40 se 59 = C

df.loc[df['Marks'] < 40, 'Grade'] = 'D'  # 40 se kam = D

print(df)
```

```python
# 163. Create score bands such as 0–100, 101–200, 201–300, and so on.

df_163 = [0, 100, 200, 300, 400, 500, 600, 700, 720]  # Marks ki range banana

df['Score_bands'] = pd.cut(df['Marks'], bins=df_163)  # Marks ko alag-alag ranges mein baantna

print(df['Score_bands'])  # Har student ka score band print karna
```

```python
# 164. Create a Score_Band column using pd.cut().

df_164 = [-100, 0, 40, 50, 600, 700]  # Marks ki alag-alag range banana

df['Marks_band'] = pd.cut(df['Marks'], bins=df_164)  # Har student ke Marks ko uski range mein rakhna

print(df['Marks_band'])  # Har student ki Marks range print karna
```

```python
# 165. Count students in each Score_Band.

df_165 = [20, 40, 60, 80, 95]  # Marks ki range banana

df['Score_Band'] = pd.cut(df['Marks'], bins=df_165)  # Har student ke Marks ko range mein baantna

df_165 = df['Score_Band'].value_counts()  # Har range mein kitne students hain, unki count nikalna

print(df_165)  # Har band  ki count print karna yahe ki har range ki 


```

```python
# 166. Create a High_Medium_Low column based on Marks.

df_166 = [0, 300, 500, 720]  # Marks ki 3 ranges banana

label = ['Low', 'Medium', 'High']  # Har range ka naam dena

df['High_Medium_Low'] = pd.cut(df['Marks'], bins=df_166, labels= label)  # Marks ke hisaab se Low, Medium, High category banana

print(df['High_Medium_Low'])  # Har student ki category print karna
```

```python
# 167. Create a Scholarship_Eligible column based on a chosen Marks threshold.

df['Scholarship_Eligible'] = 'Not_Eligible'  # Sabhi students ko pehle Not Eligible rakhna

df.loc[df['Marks'] >= 80, 'Scholarship_Eligible'] = 'Eligible'  # 80 ya usse zyada Marks walon ko Eligible karna

print(df['Scholarship_Eligible'])  # Scholarship eligibility print karna
```

```python
# 168. Create a Performance_Category column with at least four categories.

df['Performance_Category'] = " "  # Performance category ka column banaya hu 

df.loc[df['Marks'] < 35, 'Performance_Category'] = 'Poor Category'  # 35 se kam = Poor

df.loc[(df['Marks'] >= 35) & (df['Marks'] < 50), 'Performance_Category'] = 'Average Category'  # 35-49 = Average

df.loc[(df['Marks'] >= 50) & (df['Marks'] < 70), 'Performance_Category'] = 'Good Category'  # 50-69 = Good

df.loc[df['Marks'] >= 70, 'Performance_Category'] = 'Excellent Category'  # 70 ya usse zyada = Excellent

print(df['Performance_Category'])  # Category dekhna

```

```python
# 169. Calculate the numeric percentage equivalent assuming Marks are out of 720.

Marks = df['Marks']                          # Har student ke obtained Marks

Total_Marks = 720                            # Maximum Marks itna hi ho sakta hai 

Percentage = Marks / Total_Marks * 100       # Percentage calculate kr raha hu 

print(Percentage)
```

```python
# 170. Create a column indicating whether a student scored 600 or more.

df['Highest_Scored'] = 'No'  # Pehle sabhi students ko No rakhna

df.loc[df['Marks'] >= 600, 'Highest_Scored'] = 'Yes'  # 600 ya usse zyada Marks walon ko Yes karna

print(df['Highest_Scored'])  # Result dekhna
```

```python
# 171. Create a column indicating whether a student scored 650 or more.

df['Highest_Scored'] = 'No'  # Pehle sabhi students ko No rakhna

df.loc[df['Marks'] >= 650, 'Highest_Scored'] = 'Yes'  # 650 ya usse zyada Marks walon ko Yes karna

print(df['Highest_Scored'])  
```

```python
# 172. Create a column indicating whether a student scored 700 or more.

df['Highest_Scored'] = 'No'  # Pehle sabhi students ko No rakhna

df.loc[df['Marks'] >= 700, 'Highest_Scored'] = 'Yes'  # 700 ya usse zyada Marks walon ko Yes karna

print(df['Highest_Scored'])  
```

```python
# 173. Create a column indicating whether a student scored exactly 720.

df['Highest_Scored_173'] = 'No'  # Pehle maan liya ki koi student 720 nahi hai

df.loc[df['Marks'] == 720, 'Highest_Scored_173'] = 'Yes'  # Jinke Marks exactly 720 hain unko Yes karna

print(df['Highest_Scored_173'])  
```

```python
# 174. Create a column indicating whether a student scored below 200.

df['Lowest_Scored'] = 'No'  # Pehle sabhi students ko No maan liya hu 

df.loc[df['Marks'] < 200, 'Lowest_Scored'] = 'Yes'  # 200 se kam Marks walon ko Yes kr diya hu  

print(df['Lowest_Scored'])  
```

```python
# 175. Create a column containing the length of each city name.

df['City_Length'] = df['city'].str.len()  # Har city ke naam ki length nikala hu 

print(df['City_Length'])  # City ki length dekha hu 
```

```python
# 176. Create a column containing the length of each centre name.

df['center_name_Length'] = df['center_name'].str.len()  # Har centre name ki length nikala hu 

print(df['center_name_Length'])
```

```python
# 177. Create a column containing the first letter of each city.

df['City_First_Letter'] = df['city'].str[0]  # Har city ke naam ka first letter nikal kar new column mein store karna

print(df['City_First_Letter'])  # Har city ka first letter print karke result check karna
```

```python
# State ke first letter ko new column mein store karna

df['State_First_Letter'] = df['state'].str[0]  # Har state ke naam ka first letter nikal kar new column mein rakhna

print(df['State_First_Letter'])  # Har state ka first letter print karke result check karna
```

```python
# 179. Create a column combining state and city into one location label.

df['Location_Label'] = df['state'] + ' - ' + df['city']  # State aur city ko jod kar ek location banaya hu 

print(df['Location_Label'])  # Final location check karne ke liye print kiya label coloumn jo naya banaya hu usko print kara hu 
```

```python
# 180. Create a column combining city and centre name into one label.

df['Location_Label'] = df['city'] + ' - ' + df['center_name']  # City aur centre name ko jod kar ek location label banana

print(df['Location_Label'])  # Final location label check karna
```
