Level 12: Advanced Filtering & Comparative Analysis

```python
import pandas as pd
```

```python
df = pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 221. Find students scoring above the overall average Marks.

avg = df['Marks'].mean()
df_221 = df.loc[df['Marks']>avg]
print(avg)
```

```python
# 222. Find students scoring below the overall average Marks.

avg = df['Marks'].mean()
df_222 = df.loc[df['Marks'] < avg]
print(avg)
```

```python
# 223. Find students scoring equal to the maximum Marks.

maximum_Marks = df['Marks'].max()                    # Logic: Find the highest Marks using max(), then filter students scoring equal to it.
df_223 = df.loc[df['Marks']==  maximum_Marks]        # Critical Thinking: 61 students scored the maximum Marks of 720.
print(df_223)
```

```python
# 224. Find students scoring equal to the minimum Marks.

minimum_Marks = df['Marks'].min()
df_224 = df.loc[df['Marks']== minimum_Marks]
print(df_224)
```

```python
# 225. Find states whose average Marks exceed the national average.

# National average: Poore India ka average Marks
national_average = df['Marks'].mean()

# State average: Har state ka average Marks
state_average = df.groupby('state').agg({'Marks': 'mean'})

# Filter states whose average Marks exceed the national average
df_225 = state_average.loc[state_average['Marks'] > national_average]

print(df_225)

```
