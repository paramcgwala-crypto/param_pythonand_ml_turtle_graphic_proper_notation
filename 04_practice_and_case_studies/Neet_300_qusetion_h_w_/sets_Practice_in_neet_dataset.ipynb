```python
# A crime was committed at a mansion. The detective has
# 3 clues, each represented as a set of suspects which fits
# the description.

a = {"Chef", "Butler", "Maid"}       # Left handed suspects
b = {"Driver", "Butler", "Gardner"}  # Wore glasses
c = {"Butler", "Chef", "Driver"}     # Spotted in hallway


# Q1. Who is the prime suspect?
prime_suspect = a.intersection(b, c)
print("Q1. Prime suspect:", prime_suspect)


# Q2. Are the left hander suspect a subset of hallway suspect?
print("Q2. Is a subset of c?", a.issubset(c))
```
