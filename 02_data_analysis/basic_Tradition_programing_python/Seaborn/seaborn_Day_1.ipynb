```python
# Load sample dataset
df = sns.load_dataset("tips")

print(df.head())

# Create a bar chart
sns.barplot(data=df, x="size", y="total_bill")

plt.title("Bar Chart: Table Size vs Total Bill")
plt.xlabel("Number of People")
plt.ylabel("Total Bill")
plt.show()
```

```python
# Load sample dataset
df = sns.load_dataset("tips")

print(df.head())

# Create a horizontal bar chart
sns.barplot(data=df, x="total_bill", y="day")

plt.title("Horizontal Bar Chart: Day vs Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Day")
plt.show()
```

```python
sns.histplot(data=df, x="total_bill", bins=10)

plt.title("Histogram: Total Bill Distribution")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# Create a scatter plot
sns.scatterplot(data=df, x="total_bill", y="tip",
                hue="sex", size="tip")

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Create subplots
fig, ax = plt.subplots(2, 1, figsize=(12, 5))

# Line Chart
sns.lineplot(data=df, x="size", y="total_bill", ax=ax[0])
ax[0].set_title("Line Chart")
ax[0].set_xlabel("Table Size")
ax[0].set_ylabel("Total Bill")

# Bar Chart
sns.barplot(data=df, x="day", y="total_bill", ax=ax[1])
ax[1].set_title("Bar Chart")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Average Total Bill")

# Adjust spacing
plt.tight_layout()
plt.show()
```

```python
# Create two line plots in one chart
sns.lineplot(data=df, x="size", y="total_bill", label="Total Bill")
sns.lineplot(data=df, x="size", y="tip", label="Tip")

plt.title("Total Bill vs Tip")
plt.xlabel("Table Size")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
# Calculate correlation
cor = df.corr(numeric_only=True)

# Create heatmap
sns.heatmap(cor, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```
