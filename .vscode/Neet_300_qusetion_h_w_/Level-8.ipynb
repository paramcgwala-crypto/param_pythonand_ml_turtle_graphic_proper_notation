```python
import pandas as pd 
```

```python
df =pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 141. Use groupby().agg() to calculate count, mean, maximum, and minimum Marks by state.

df1 = df.groupby('state').agg({'Marks': ['count', 'mean', 'max', 'min']})
print(df1)

```

```python
# 142. Use groupby().agg() to calculate count, mean, and standard deviation by city.

df2 = df.groupby('city').agg({'Marks':['count','mean','std']})
print(df2)
```

```python
# 143. Use groupby().agg() to calculate total and average Marks by centre.

df3 = df.groupby('center_name').agg({'Marks':['sum','mean']})
print(df3)
```

```python
# 144. Give meaningful names to aggregated columns for state-level statistics.

df4 = df.groupby('state').agg(Total_Students = ('Marks','count'), Total_Average = ('Marks','min'))
print(df4)
```

```python
# 145. Calculate multiple aggregations for Marks by state in one operation.

df5 = df.groupby('state').agg({'Marks':['count','max','mean','min','median','sum']})
print(df5)
```

```python
# 146. Calculate the number of students and average Marks for each city.

df6 = df.groupby('city').agg({'Marks':['count','mean']})
print(df6)
```

```python
# 147. Calculate the number of students, total Marks, and average Marks for each centre.
df7 = df.groupby('center_name').agg({'Marks':['count','mean','sum']})   # eaqch center matlb hindi mai hoga har center 
print(df7)
```
