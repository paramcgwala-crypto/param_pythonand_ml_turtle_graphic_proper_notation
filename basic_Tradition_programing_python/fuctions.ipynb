```python
def quad(a, b, c):
    x_1 = (-b + (b**2-4*a*c)**0.5)/2*a
    x_2 = (-b - (b**2-4*a*c)**0.5)/2*a
    return (x_1, x_2)
```

```python
def fact(n):
    f = 1
    for i in range(1, n+1):
        f = f * i
    return f

fact(5)
```

```python
def fact(n):  # Factorial ka function banaya
    a = 1  # Multiplication ki starting value
    for i in range(n, 1, -1):  # n se 2 tak ulta loop
        a = a * i  # Har number ko multiply kiya
    return a  # Final factorial return kiya

fact(5)  # 5 ka factorial nikala
```

```python
# WAP to check whether a number is prime or not

def prime(n):
    if n < 2:
        return False  # 2 se chhota prime nahi hota

    for i in range(2, n):
        if n % i == 0:
            return False  # Puri tarah divide ho gaya

    return True  # Koi divisor nahi mila

prime(7)
```

```python
# WAP to reverse a number

def rev(n):
    a = 0
    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao
    return a

rev(9212394109)
```

```python
# WAP to check whether a number is palindrome or not

def pal(n):
    original = n
    a = 0

    while n > 0:
        r = n % 10  # Last digit nikalo
        a = a * 10 + r  # Reverse number banao
        n = n // 10  # Last digit hatao

    return original == a  # Original aur reverse compare karo

pal(121)
```

```python
# Question 3: Sum of Digits
# Task: Python mein def function banao jo kisi number ke sabhi digits ko add karke return kare.

def sum_digits(n):  # Digits ka sum nikalne ka function
    a = 0  # Total sum store karne ke liye

    while n > 0:  # Jab tak number 0 se bada hai
        b = n % 10  # Number ka last digit nikalo
        a = a + b  # Last digit ko total sum mein add karo
        n = n // 10  # Number ka last digit hatao

    return a  # Final sum return karo

sum_digits(325)  # 3 + 2 + 5 = 10
```

```python
# chalo Question 4: Count Digits — ab tu khud logic lagana! 

def count_digits(n):  # Number ke digits count karne ka function
    count = 0  # Digits ka count 0 se start kiya

    while n > 0:  # Jab tak number 0 se bada hai
        count = count + 1  # Har digit ke liye count 1 badhao
        n = n // 10  # Number ka last digit hatao

    return count  # Total digits return karo

count_digits(325)  # Output: 3

```

```python
# Question 5: Largest of Two Numbers
# Task: Python mein def function banao jo do numbers mein se bada number return kare.

def largest(a, b):  # Do numbers mein se bada number nikalne ka function

    if a > b:  # Check karo kya a, b se bada hai
        return a  # Agar haan, toh a return karo
    else:  # Agar a bada nahi hai
        return b  # Toh b return karo

print(largest(10, 25))  # Output: 25
print(largest(50, 20))  # Output: 50
```

```python
# 🐍 Question 7: Even or Odd
# Task: Ek Python function banao jo check kare ki diya gaya number even hai ya odd.
# Examples:
# print(check_number(10))  # Output: Evenprint(check_number(7))   # Output: Oddprint(check_number(0))   # Output: Even



# Rules:
# - def use karna hai.
# - if-else use karna hai.
# - return use karna hai.
# - % operator se remainder check karna hai.

def ceck_even_odd(n):  # Number check karne ka function
    if n % 2 == 0:     # Remainder 0 hai toh even
        return "Even"  # Even return karo
    else:              # Warna odd hoga
        return "Odd"   # Odd return karo

print(ceck_even_odd(8))  # Output: Even
```

```python
# 🔥 Question 8: Positive, Negative ya Zero
# Task: def function banao jo check kare ki number positive hai, negative hai ya zero.
# Examples:
# print(check_number(10))   # Positiveprint(check_number(-5))   # Negativeprint(check_number(0))    # Zero

def positive_negative_check(n):
    if n > 0:
        return 'Positive'
    elif n < 0:
        return 'Negative'
    else:
        return 'Zero'

print(positive_negative_check(10))   # Positive
print(positive_negative_check(-5))   # Negative
print(positive_negative_check(0))    # Zero
```

```python
# 🔥 Question 9: Largest of Three Numbers
# Task: def function banao jo teen numbers mein se sabse bada number return kare.
# Examples:

def Largest_of_3(a, b, c):  # Teen numbers mein se largest nikalne ka function
    if a > b and a > c:    # Check karo a, dono se bada hai ya nahi
        return a             # Agar haan, toh a return karo
    elif b > a and b > c:  # Check karo b, dono se bada hai ya nahi
        return b             # Agar haan, toh b return karo
    else:                    # Agar a aur b largest nahi hain
        return c             # Toh c return karo

print(Largest_of_3(10, 3, 2))   # Output: 10
print(Largest_of_3(4, 15, 9))   # Output: 15
print(Largest_of_3(7, 12, 20))  # Output: 20
```

```python
# 🐍 Question 10: Sum of Digits
# # Task: Python mein ek function banao jo kisi number ke saare digits ka sum return kare.
# # Examples:

# def sum_of_digit(n):
#     a = 0 

#     while n > 0:
#         n %  10
#         n // 10
#     return a

def sum_of_digit(n):  # Digits ka sum nikalne ka function
    a = 0              # Sum ko 0 se start kar rahe hain

    while n > 0:       # Jab tak number 0 se bada hai
        digit = n % 10 # Last digit nikaalo
        a = a + digit  # Digit ko sum mein add karo
        n = n // 10    # Last digit hata do

    return a           # Final sum return karo

print(sum_of_digit(1234))  # Output: 10
print(sum_of_digit(567))   # Output: 18
print(sum_of_digit(90))    # Output: 9
```

```python
# 🐍 Question 11: Reverse a Number
# Task: Python mein ek function banao jo diye gaye number ko reverse karke return kare.
# Examples:
# print(reverse_number(1234))  # Output: 4321print(reverse_number(567))   # Output: 765print(reverse_number(120))   # Output: 21



# Rules:
# - def aur return compulsory.
# - while loop use karna hai.
# - % 10 aur // 10 use karna hai.
# - Number ko string mein convert nahi karna.

def Reverse_a_Number(n):  # Number ko reverse karne ka function
    a = 0                 # Reverse number ko 0 se start kiya

    while n > 0:          # Jab tak number 0 se bada hai
        digit = n % 10    # Last digit nikaalo
        a = a * 10 + digit  # Reverse number mein digit jodo
        n = n // 10       # Original number se last digit hatao

    return a              # Final reverse number return karo

print(Reverse_a_Number(1234))  # Output: 4321
print(Reverse_a_Number(567))   # Output: 765
print(Reverse_a_Number(120))   # Output: 21
```
