```python
A = set(range(4, 101, 4))
B = set(range(6, 101, 6))

print("A =", sorted(A))
print("B =", sorted(B))

print("A union B =", sorted(A | B))
print("A intersection B =", sorted(A & B))

print("Only divisible by 4 =", sorted(A - B))
print("Only divisible by 6 =", sorted(B - A))

print("Divisible by both 4 and 6 =", sorted(A & B))
```
