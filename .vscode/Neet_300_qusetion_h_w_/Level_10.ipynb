```python
import pandas as pd 
```

```python
df =pd.read_csv('all_neet_marks_2607.csv')
```

```python
# 181 Convert state names to uppercase

df['state'] = df['state'].str.upper()
print(df)
```

```python
# 182 Convert city names to lowercase.

df['city'] = df['city'].str.lower()

print(df)
```

```python
# 183 Remove leading and trailing spaces from state values.

df['state']= df['state'].str.strip()
print(df)
```

```python
# 184 Remove leading and trailing spaces from city values.
df['city']= df['city'].str.strip()
print(df)
```

```python
# 185 Find rows where city contains a specific word, ignoring case.

df.loc[df['city'].str.contains('R'),'Marks'].mean()    #strip() ka matlab hota hai string ke starting aur ending ke extra spaces ko hataana.
```

```python
# 186 Find rows where center_name contains 'school', ignoring case.

df.loc[df['center_name'].str.contains('school')]
```

```python
# 187 Extract the first word from center_name.

df_187 = df['center_name'].str.split().str[0]      # center_name column ko select kar rahe hain
                                                   # split() har center_name ko words me tod raha hai
                                                   # str[0] har row ka first word nikal raha hai
print(df_187)
```

```python
# 188 Extract the last word from center_name.

df_188 = df['center_name'].str.split().str[-1]

print(df_188)
```

```python
# 189 Find the length of each state name.

df_189 = df['state'].str.len()     # state column ko select karke, har state name ke characters ki length nikal rahe hain
print(df_189)
```

```python
# 190 Find the length of each city name.

df_190 = df['city'].str.len()
print(df_190)
```

```python
# 191 Find the length of each centre name.


df_191 = df['center_name'].str.len()
print(df_191 )
```

```python
# 192 Replace a selected word in center_name with another word.

df_192 = df['center_name'].str.replace( 'KENDRIYA' , 'kENDARIYA_vIDHAYALAYA')
print(df_192)
```

```python
# 193 Replace multiple variations of a centre name with a standard value.

df_193 = df['center_name'].str.replace('KENDRIYA','Vidhayalaye')
df_193 = df_193.str.replace('CHAKDAHA','Vidhayalaye')
print(df_193)
```

```python
# 194 Check whether city values start with a particular prefix.

df_194 = df['city'].str.startswith('Ra')   # prefix yani starting ke word  
print(df_194)
```

```python
# 195 Check whether city values end with a particular suffix.

df_195 = df['city'].str.endswith('pur')   # suffix yani endd ke word 
print(df_195)
```

```python
# 196 Split center_name into separate words.

df_196 = df['center_name'].str.split()      # center_name ko alag-alag words me tod raha hai
print(df_196)
```

```python
# 197 Extract a specific part of center_name using string slicing.

df_197 = df['center_name'].str[0:4]        # center_name ke 0 se 4 index tak ke characters nikal raha hai

print(df_197)
```

```python
# 198 Find the number of spaces in each center_name.

df_198 = df['center_name'].str.count(' ')      # har center_name me spaces count kar raha hai
print(df_198)                                  #mtlab har index mai hamra spaces zero hai ye spaces ka vount dekhnre ko bola gaya hai qquestion mai sir ne bola hai brother
```

```python
# 199 Create a standardized version of state and city names.

df_199 = df[['state','city']]    # state aur city columns ko select karke ek naya DataFrame bana rahe hain
df_199 = df['state'].str.upper() # state ke sabhi names ko uppercase me convert kar rahe hain
df_199 = df['city'].str.upper()  # city ke sabhi names ko uppercase me convert kar rahe hain
print(df_199)                    # standardized state aur city ko print kar rahe hain
```

```python
# 200 Find duplicate centre names after converting them to lowercase.

df_200 = df['center_name'].str.lower().duplicated()      # center_name ko lowercase me convert karke duplicate values check kar rahe hain

print(df_200)
```
