```python
a = ['apple', 'banana', 'apple', 'banana', 'grapes']

d = {}

for i in a:
    v = 0
    c = 0

    for j in i:
        if j in 'aeiou':
            v += 1
        else:
            c += 1

    d[i] = {'Vowel': v, 'consonant': c}

print(d)
```

```python
a = ['apple', 'banana', 'apple', 'banana', 'grapes']

d = {}

for i in a:
    f = {}

    for j in i:
        if j in f:
            f[j] += 1
        else:
            f[j] = 1

    d[i] = f

print(d)
```
