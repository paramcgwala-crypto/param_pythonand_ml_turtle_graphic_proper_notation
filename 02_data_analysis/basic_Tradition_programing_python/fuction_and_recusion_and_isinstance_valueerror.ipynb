```python
# <!-- Practice Question 1 — Student Marks Validator
# Ek function check_marks(name, marks) banao jo:
# 1. Check kare ki name string hai.
# 2. Check kare ki marks ek list hai.
# 3. List ke har element ka type int ya float hai. True aur False allowed nahi hain.
# 4. Agar koi data type galat ho, toh raise TypeError use kare.
# 5. Agar saara data sahi ho, toh marks ka total return kare. -->


def ceck_marks(name, marks):  # Function bana rahe hain, jo name aur marks lega

    if not isinstance(name, str):  # Check kar rahe hain ki name string hai ya nahi
        raise TypeError("Name string hona chahiye")  # Galat type hua toh error aayega

    if not isinstance(marks, list):  # Check kar rahe hain ki marks ek list hai ya nahi
        raise TypeError("Marks list honi chahiye")  # List nahi hui toh error aayega

    total = 0  # Total ko 0 se start kar rahe hain

    for i in marks:  # List ke har mark ko ek-ek karke i mein lenge

        if isinstance(i, bool):  # Check kar rahe hain ki value True ya False toh nahi
            raise TypeError("True aur False allowed nahi hain")  # Boolean hua toh error aayega

        if not isinstance(i, (int, float)):  # Check kar rahe hain ki mark int ya float hai
            raise TypeError("Marks sirf int ya float hone chahiye")  # Galat type hua toh error

        total = total + i  # Har valid mark ko total mein add kar rahe hain

    return total  #Saare marks add hone ke baad total return kar rahe hain


print(ceck_marks("Param", [80, 75, 90]))  #Function call karke result print kar rahe hain
```

```python

# Practice Question 2 - Number Checker

# Ek function check_number(num) banao.

# 1. Check karo ki num ka type int hai.
# 2. Agar type galat ho, toh raise TypeError use karo.
# 3. True aur False ko allow nahi karna.
# 4. Agar sab sahi hai, toh return num * 2 karo.

# Example:
# print(check_number(5))

# Expected Output:
# 10

def check_number(num):  # Function bana rahe hain jo num naam ka input lega

    # Pehle check kar rahe hain ki num True ya False toh nahi hai
    if isinstance(num, bool):
        raise TypeError("Boolean values allowed nahi hain")  # Bool hua toh error dekar rok do

    # Check kar rahe hain ki num integer type ka hai ya nahi
    if not isinstance(num, int):
        raise TypeError("num sirf int hona chahiye")  # Int nahi hua toh error de do

    # Agar dono checks pass ho gaye, toh number ko 2 se multiply karke return karo
    return num * 2


# Function ko 5 dekar call kar rahe hain aur result print kar rahe hain
print(check_number(5))  # Output: 10
```

```python
# Q1. Calculate Total Marks
# Task: calculate_total(marks) naam ka function bana.
# Conditions:
# 1. marks ek list honi chahiye.
# 2. Agar list nahi hai, toh raise TypeError use karna.
# 3. for loop se list ke har element ko check karna.
# 4. Har element int ya float hona chahiye. True aur False allowed nahi hain.
# 5. Galat element par raise TypeError karna.
# 6. Saare marks ka total return karna.

def calculate_total(marks):  # Function banaya jo marks ki list lega

    # Check kar rahe hain ki marks ek list hai ya nahi
    if not isinstance(marks, list):
        raise TypeError("marks must be a list")  # Galat type par error

    total = 0  # Total ko 0 se start kiya

    # List ke har element ko ek-ek karke check karenge
    for i in marks:

        # Pehle Boolean check karna zaroori hai
        # Kyunki Python mein bool, int ka subtype hai
        if isinstance(i, bool):
            raise TypeError("Boolean values are not allowed")

        # Check kar rahe hain ki element int ya float hai
        if not isinstance(i, (int, float)):
            raise TypeError("Marks must contain only int or float")

        # Current number ko total mein add kar rahe hain
        total = total + i

    # Final total ko function se return kar rahe hain
    return total


# Function ko call karke result check kar rahe hain
print(calculate_total([80, 50, 60]))  # Output: 190
```

```python
# Q2. Count Even Numbers

# Task: count_even(numbers) naam ka function banao.

# Conditions:

# numbers ek list honi chahiye, warna TypeError raise karo.

# List ke har element ko for loop se check karo.

# Har element integer (int) hona chahiye.

# float, string, True aur False allowed nahi hain.

# List mein jitne even numbers hain, unki counting karke return karo.

def count_even(numbers):

    # Check kar rahe hain ki numbers ek list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError("numbers must be a list")

    # Even numbers ki counting 0 se start karenge
    total = 0

    # List ke har element ko check karenge
    for i in numbers:

        # Boolean values True aur False ko allow nahi karna
        if isinstance(i, bool):
            raise TypeError("Boolean values are not allowed")

        # Check kar rahe hain ki har element integer hai ya nahi
        if not isinstance(i, int):
            raise TypeError("Only integers are allowed")

        # Check kar rahe hain ki number even hai ya nahi
        if i % 2 == 0:
            total = total + 1  # Even number milne par count 1 badhayenge

    # Total even numbers ka count return karenge
    return total


# Function ko test kar rahe hain
print(count_even([2, 5, 8, 11, 14]))  # Output: 3
print(count_even([1, 3, 5, 7]))       # Output: 0
print(count_even([30, 60, 90]))       # Output: 3
```

```python
# Q3. Count Positive Numbers

# Task: count_positive(numbers) naam ka function banao.

# Conditions:

# numbers ek list honi chahiye, warna TypeError raise karo.

# List ke har element ko for loop se check karo.

# Har element int ya float hona chahiye.

# True aur False allowed nahi hain.

# Agar koi element string ya doosre invalid type ka hai, toh TypeError raise karo.

# Sirf positive numbers count karo aur total count ko return karo.

def count_positive(numbers):
    # Check kar rahe hain ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError('Its only allowed list')

    # Positive numbers ko count karne ke liye 0 se start
    count = 0

    # List ke har element ko check karenge
    for i in numbers:

        # Check kar rahe hain ki element int ya float hai
        if not isinstance(i, (int, float)):
            raise TypeError('Its only allowed int and float')

        # True aur False ko allow nahi karna
        # Kyunki Python mein bool, int ka subtype hai
        if isinstance(i, bool):
            raise TypeError('Boolean values are not allowed')

        # Agar number 0 se bada hai, toh positive hai
        if i > 0:
            # Positive number milne par count 1 badha do
            count = count + 1

    # Loop complete hone ke baad final count return karo
    return count


# Function ko call karke check kar rahe hain
print(count_positive([20, 30, 60, 90]))
```

```python
# Q4 — count_negative(numbers)
# Ek function bana jo list ke andar negative numbers ko count kare.
# Rules:
# 1. Function ka naam count_negative(numbers) hona chahiye.
# 2. Agar input list nahi hai, toh TypeError raise karna.
# 3. List ke elements sirf int ya float hone chahiye.
# 4. True aur False allowed nahi hain.
# 5. Negative numbers ko count karna hai, unka sum nahi.
# 6. return loop ke bahar hona chahiye.

def count_negative(numbers):

    # Check kar rahe hain ki input list hai ya nahi
    if not isinstance(numbers, list):
        raise TypeError('Its input allowed a list')

    # Negative numbers ko count karne ke liye 0 se start
    count = 0

    # List ke har element ko check karenge
    for i in numbers:

        # True aur False ko allow nahi karna
        # Kyunki Python mein bool, int ka subtype hai
        if isinstance(i, bool):
            raise TypeError('True and False are not allowed')

        # Check kar rahe hain ki element int ya float hai
        if not isinstance(i, (int, float)):
            raise TypeError('Its allowed to use int and float')

        # Agar number 0 se chhota hai, toh negative hai
        if i < 0:
            # Negative number milne par count 1 badha do
            count = count + 1

    # Loop complete hone ke baad final count return karo
    return count


# Function ko call karke test kar rahe hain
print(count_negative([39, 33, -5, -9]))
```
