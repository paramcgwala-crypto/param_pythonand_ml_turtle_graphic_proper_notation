```python
import pandas as pd 
```

```python
df =pd.read_csv('all_neet_marks_2607.csv')
```

```python
# Find all the cities that has performed better than their state avg

# 1. df1 state, city groupby marks avg
# 2. df2 state groupby marks avg
# 3. Merge df1 and df2
# """

df1 = df.groupby(['state','city']).agg({'Marks':'mean'}).reset_index()
df2 = df.groupby('state').agg({'Marks':'mean'}).reset_index()

a = df1.merge(df2,on= 'state',how = 'inner')
print(a.loc[a['Marks_x']>a['Marks_y'],])

```

```python
# """
# Find all the centers that has performed better than their city avg
# """

df3 = df.groupby(['city','center_num']).agg({'Marks':'mean'}).reset_index()

df4 = df.groupby('city').agg({'Marks':'mean'}).reset_index()

b = df3.merge(df4, on='city', how='inner')

print(b.loc[b['Marks_x'] > b['Marks_y'],])
```

```python
# """
# Find all the centers that has performed better than their city avg
# and state avg
# """

df5= df.groupby(['state','city','center_num']).agg({'Marks':'mean'}).reset_index()

df6 = df.groupby(['state','city']).agg({'Marks':'mean'}).reset_index()

df7 = df.groupby(['state']).agg({'Marks':'mean'}).reset_index()

c = df5.merge(df6, on=['state','city'], how='inner')
c = c.merge(df7,on= 'state', how= 'inner')
print(c.loc[(c['Marks_x'] > c['Marks_y']) & (c['Marks_x'] > c['Marks']),])
```
