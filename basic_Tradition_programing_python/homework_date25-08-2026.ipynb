```python
# Q1. Count the frequency of each word in the list.
# a = ['apple','banana','apple','banana','grapes']
# o/p
# {'apple':2,
#  'banana':2,
#  'grapes':1}

a = ['apple','banana','apple','banana','grapes']
count = {}

for i in a:
    if i in count:
        count[i] = count[i] + 1
    else:
        count[i] = 1

print("frequency of word hai ye =",count)
```
