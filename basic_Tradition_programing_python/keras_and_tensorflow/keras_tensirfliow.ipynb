```python
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
```

```python
x1 = np.array([0.2, 0.4, 0.6, 0.8, 0.9])
x2 = np.array([0.8, 0.7, 0.4, 0.3, 0.6])   #input
```

```python
X = np.column_stack((x1, x2))
y = np.array([0, 0, 1, 1, 1])
```

```python
print("Features (X):\n", X)
print("Targets (y):\n", y)
```

```python
model = keras.Sequential([
    layers.Input(shape=(2,)),  # 2 input nodes
    layers.Dense(4, activation='relu'),
    layers.Dense(1, activation='sigmoid')  # 1 output node
])
```

```python
model.compile(optimizer=keras.optimizers.Adam(learning_rate=0.05),
              loss='mean_squared_error',
              metrics=['accuracy'])
```

```python
history = model.fit(X, y, epochs=10, verbose=1)

# Evaluate and display predictions
predictions = model.predict(X)
print("\nRaw Predictions:\n", predictions)
print("\nRounded Predictions (0 or 1):\n", predictions.round())
```

```python
history = model.fit(X, y, epochs=100, verbose=1)

# Evaluate and display predictions
predictions = model.predict(X)
print("\nRaw Predictions:\n", predictions)
print("\nRounded Predictions (0 or 1):\n", predictions.round())
```

```python
history = model.fit(X, y, epochs=200, verbose=1)

# Evaluate and display predictions
predictions = model.predict(X)
print("\nRaw Predictions:\n", predictions)
print("\nRounded Predictions (0 or 1):\n", predictions.round())
```

```python
from sklearn.metrics import precision_score, recall_score

history = model.fit(X, y, epochs=100, verbose=1)

# Evaluate and display predictions
predictions = model.predict(X)
print("\nRaw Predictions:\n", predictions)
print("\nRounded Predictions (0 or 1):\n", predictions.round())

# Precision and Recall
y_pred = (predictions > 0.5).astype(int).flatten()

print("\nPrecision:", precision_score(y, y_pred))
print("Recall:", recall_score(y, y_pred))
```
