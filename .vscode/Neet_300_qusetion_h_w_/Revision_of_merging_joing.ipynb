Merging concept basics revision

```python
import pandas as pd

employees = pd.DataFrame({
    "emp_id": [1, 2, 3, 4, 5],
    "name": ["Aman", "Riya", "Karan", "Simran", "Param"],
    "dept_id": [10, 20, 10, 30, 40]
})

departments = pd.DataFrame({
    "dept_id": [10, 20, 30, 50],
    "dept_name": ["IT", "HR", "Sales", "Finance"]
})

salaries = pd.DataFrame({
    "emp_id": [1, 2, 3, 6],
    "salary": [50000, 45000, 60000, 70000]
})

# Q1. Inner join: sirf common rows
pd.merge(employees, departments, on="dept_id", how="inner")

# Q2. Left join: employees ki saari rows, match nahi to NaN
pd.merge(employees, departments, on="dept_id", how="left")

# Q3. Outer join: dono ki saari rows, _merge se source pata chalega
pd.merge(employees, departments, on="dept_id", how="outer", indicator=True)

# Q4. 3 tables ek saath: pehle dept, phir salary jodo
(employees
    .merge(departments, on="dept_id", how="left")   # dept ka naam
    .merge(salaries, on="emp_id", how="left"))      # salary


# Q5. Key ka naam alag ho to left_on / right_on
salaries2 = salaries.rename(columns={"emp_id": "eid"})
pd.merge(employees, salaries2, left_on="emp_id", right_on="eid").drop(columns="eid")  # eid extra tha, hata diya
```
